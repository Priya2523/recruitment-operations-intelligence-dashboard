# Candidate Reconciliation Analysis

```python
# ============================================
# STEP 9: CANDIDATE RECONCILIATION ANALYSIS

# ============================================

# We already loaded:
# candidate_df  -> 814-record cleaned master
# shortlist_df  -> 59 shortlisted candidates

# ------------------------------------------------
# 1. Create normalized comparison keys
# ------------------------------------------------

candidate_df["_name_key"] = (
    candidate_df["candidate_name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

shortlist_df["_name_key"] = (
    shortlist_df["Candidate Name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)


# ------------------------------------------------
# 2. Find duplicate names in the 814-record master
# ------------------------------------------------

candidate_name_counts = (
    candidate_df[
        candidate_df["_name_key"] != ""
    ]["_name_key"]
    .value_counts()
)

duplicate_names = candidate_name_counts[
    candidate_name_counts > 1
].index


# ------------------------------------------------
# 3. Build reconciliation report
# ------------------------------------------------

reconciliation_rows = []

for _, row in shortlist_df.iterrows():

    name_key = row["_name_key"]

    if not name_key:
        continue

    matches = candidate_df[
        candidate_df["_name_key"] == name_key
    ]

    if len(matches) == 0:

        reconciliation_rows.append({
            "shortlist_name": row["Candidate Name"],
            "shortlist_role": row["Position Applied"],
            "recruiter": row["Recuriter Name"],
            "match_status": "NO_NAME_MATCH",
            "master_match_count": 0,
            "master_role": None,
            "master_date": None,
            "review_required": "YES"
        })

    else:

        for _, match in matches.iterrows():

            master_role = match.get("position_applied", "")

            reconciliation_rows.append({
                "shortlist_name": row["Candidate Name"],
                "shortlist_role": row["Position Applied"],
                "recruiter": row["Recuriter Name"],
                "match_status": "NAME_MATCH",
                "master_match_count": len(matches),
                "master_role": master_role,
                "master_date": match.get("date", None),
                "review_required": (
                    "YES"
                    if len(matches) > 1
                    else "REVIEW ROLE"
                )
            })


reconciliation_df = pd.DataFrame(reconciliation_rows)


# ------------------------------------------------
# 4. Classify potential matches
# ------------------------------------------------

def classify_match(row):

    if row["match_status"] == "NO_NAME_MATCH":
        return "UNMATCHED"

    if row["master_match_count"] > 1:
        return "DUPLICATE NAME - REVIEW"

    shortlist_role = str(row["shortlist_role"]).lower()
    master_role = str(row["master_role"]).lower()

    # Basic role-word overlap for investigation only
    shortlist_words = set(
        shortlist_role.replace("-", " ").split()
    )

    master_words = set(
        master_role.replace("-", " ").split()
    )

    common_words = shortlist_words.intersection(master_words)

    if len(common_words) >= 2:
        return "NAME + ROLE SUPPORT"

    return "NAME MATCH - ROLE REVIEW"


reconciliation_df["match_classification"] = (
    reconciliation_df.apply(
        classify_match,
        axis=1
    )
)


# ------------------------------------------------
# 5. Summary
# ------------------------------------------------

print("========== RECONCILIATION SUMMARY ==========")

print(
    "Shortlist records analysed:",
    len(shortlist_df)
)

print(
    "Potential name-linked records:",
    (reconciliation_df["match_status"] == "NAME_MATCH").sum()
)

print(
    "Unmatched records:",
    (reconciliation_df["match_status"] == "NO_NAME_MATCH").sum()
)

print("\nClassification:")
print(
    reconciliation_df["match_classification"]
    .value_counts()
)


# ------------------------------------------------
# 6. Display reconciliation report
# ------------------------------------------------

print("\n========== RECONCILIATION REPORT ==========")

display(
    reconciliation_df[
        [
            "shortlist_name",
            "shortlist_role",
            "recruiter",
            "match_classification",
            "master_role",
            "master_date",
            "master_match_count"
        ]
    ]
)
```

# ROBUST INTERVIEW DATE PARSER

```python
# ============================================================
# STEP 11.2 — ROBUST INTERVIEW DATE PARSER
# ============================================================

import pandas as pd
import re
from datetime import datetime

print("=" * 60)
print("STEP 11.2 — ROBUST INTERVIEW DATE PARSER")
print("=" * 60)


# ------------------------------------------------------------
# 1. ORIGINAL COLUMN → CLEAN DATE TEXT
# ------------------------------------------------------------

def clean_date_text(value):
    if pd.isna(value):
        return None

    text = str(value).strip()

    if not text or text.lower() in ["nan", "none", "nat"]:
        return None

    # Normalize spaces
    text = re.sub(r"\s+", " ", text)

    # Normalize common apostrophe variations
    text = text.replace("’", "'").replace("`", "'")

    return text


# ------------------------------------------------------------
# 2. ROBUST DATE PARSER
# ------------------------------------------------------------

def parse_interview_date(value):

    text = clean_date_text(value)

    if text is None:
        return pd.NaT

    # --------------------------------------------------------
    # CASE 1:
    # Explicit ISO datetime/date
    # Example: 2026-05-04 00:00:00
    # IMPORTANT: YYYY-MM-DD MUST NOT use dayfirst=True
    # --------------------------------------------------------

    iso_match = re.search(
        r"\b(20\d{2})-(\d{1,2})-(\d{1,2})\b",
        text
    )

    if iso_match:
        year, month, day = map(int, iso_match.groups())

        try:
            return pd.Timestamp(year=year, month=month, day=day)
        except ValueError:
            pass


    # --------------------------------------------------------
    # CASE 2:
    # Rescheduled interview
    #
    # Example:
    # 06-05-2026 ( rescheduled ) 08-05-2026
    #
    # Business rule:
    # Use the FINAL / RESCHEDULED date.
    # --------------------------------------------------------

    rescheduled_dates = re.findall(
        r"\b(\d{1,2})[-/](\d{1,2})[-/](20\d{2})\b",
        text
    )

    if "reschedul" in text.lower() and len(rescheduled_dates) >= 2:

        day, month, year = map(int, rescheduled_dates[-1])

        try:
            return pd.Timestamp(
                year=year,
                month=month,
                day=day
            )
        except ValueError:
            pass


    # --------------------------------------------------------
    # CASE 3:
    # DD-MM-YY / DD-MM-YYYY
    #
    # Examples:
    # 28-04-26
    # 14-4-26
    # 22 -05- 2026
    # --------------------------------------------------------

    numeric_date = re.search(
        r"\b(\d{1,2})\s*[-/]\s*(\d{1,2})\s*[-/]\s*(\d{2,4})\b",
        text
    )

    if numeric_date:

        day, month, year = numeric_date.groups()

        day = int(day)
        month = int(month)
        year = int(year)

        if year < 100:
            year += 2000

        try:
            return pd.Timestamp(
                year=year,
                month=month,
                day=day
            )
        except ValueError:
            pass


    # --------------------------------------------------------
    # CASE 4:
    # Month-name dates
    #
    # Examples:
    # May 1'st 4pm fine
    # May 1st 3 pm
    # May 4
    # --------------------------------------------------------

    month_pattern = (
        r"\b("
        r"Jan(?:uary)?|"
        r"Feb(?:ruary)?|"
        r"Mar(?:ch)?|"
        r"Apr(?:il)?|"
        r"May|"
        r"Jun(?:e)?|"
        r"Jul(?:y)?|"
        r"Aug(?:ust)?|"
        r"Sep(?:tember)?|"
        r"Oct(?:ober)?|"
        r"Nov(?:ember)?|"
        r"Dec(?:ember)?"
        r")"
        r"\s+"
        r"(\d{1,2})"
    )

    month_match = re.search(
        month_pattern,
        text,
        flags=re.IGNORECASE
    )

    if month_match:

        month_name = month_match.group(1)
        day = int(month_match.group(2))

        # Internship shortlist dates are from 2026
        year = 2026

        try:
            parsed = pd.to_datetime(
                f"{month_name} {day} {year}",
                format="%B %d %Y"
            )

            return parsed

        except Exception:

            try:
                parsed = pd.to_datetime(
                    f"{month_name} {day} {year}",
                    format="%b %d %Y"
                )

                return parsed

            except Exception:
                pass


    # --------------------------------------------------------
    # CASE 5:
    # Fallback — only for genuinely unambiguous dates
    # --------------------------------------------------------

    # Do NOT use dayfirst=True globally.
    # This fallback is deliberately conservative.

    return pd.NaT


# ------------------------------------------------------------
# 3. APPLY TO SHORTLIST DATA
# ------------------------------------------------------------

shortlist_df["_interview_date_original"] = shortlist_df["Interview Date"]

shortlist_df["parsed_interview_date_v2"] = (
    shortlist_df["Interview Date"]
    .apply(parse_interview_date)
)


# ------------------------------------------------------------
# 4. DATE STATUS
# ------------------------------------------------------------

def classify_date_status(row):

    original = row["Interview Date"]
    parsed = row["parsed_interview_date_v2"]

    if pd.isna(original) or str(original).strip().lower() in ["nan", "none", ""]:
        return "MISSING"

    if pd.isna(parsed):
        return "UNPARSED - REVIEW"

    return "PARSED"


shortlist_df["date_status_v2"] = (
    shortlist_df.apply(classify_date_status, axis=1)
)


# ------------------------------------------------------------
# 5. SUMMARY
# ------------------------------------------------------------

print("\n========== DATE STATUS SUMMARY ==========\n")

print(
    shortlist_df["date_status_v2"]
    .value_counts()
)


# ------------------------------------------------------------
# 6. COVERAGE
# ------------------------------------------------------------

total_records = len(shortlist_df)

parsed_count = (
    shortlist_df["date_status_v2"] == "PARSED"
).sum()

missing_count = (
    shortlist_df["date_status_v2"] == "MISSING"
).sum()

unparsed_count = (
    shortlist_df["date_status_v2"] == "UNPARSED - REVIEW"
).sum()

coverage = (
    parsed_count / total_records * 100
    if total_records > 0 else 0
)

print("\n========== DATE COVERAGE ==========")

print(f"Total shortlist records : {total_records}")
print(f"Successfully parsed     : {parsed_count}")
print(f"Missing interview date  : {missing_count}")
print(f"Unparsed / review       : {unparsed_count}")
print(f"Date coverage           : {coverage:.1f}%")


# ------------------------------------------------------------
# 7. SHOW ORIGINAL VS PARSED
# ------------------------------------------------------------

display_columns = [
    "Candidate Name",
    "Position Applied",
    "Interview Date",
    "parsed_interview_date_v2",
    "date_status_v2"
]

print("\n========== DATE PARSING RESULTS ==========\n")

display(
    shortlist_df[display_columns]
)


# ------------------------------------------------------------
# 8. SHOW ONLY RECORDS REQUIRING REVIEW
# ------------------------------------------------------------

review_df = shortlist_df[
    shortlist_df["date_status_v2"] != "PARSED"
][display_columns]

print("\n========== DATES REQUIRING REVIEW ==========\n")

display(review_df)


# ------------------------------------------------------------
# 9. VALIDATION OF IMPORTANT KNOWN CASES
# ------------------------------------------------------------

print("\n========== KNOWN DATE VALIDATION ==========\n")

known_candidates = [
    "Vishwa Patel",
    "Anjali Thakkar",
    "Prasad Pawar",
    "Sanket Lunawat",
    "Amit Kewal",
    "Arpit Shah",
    "Abhishek Bhatia",
    "Himanshu Wadhwa",
    "Sreenu Rayavarapu",
    "LOKESH NISHANTH",
    "Anusha Chimata",
    "MANU GOYAL",
    "Supriya Awasthi"
]

validation_df = shortlist_df[
    shortlist_df["Candidate Name"]
    .astype(str)
    .isin(known_candidates)
][
    [
        "Candidate Name",
        "Interview Date",
        "parsed_interview_date_v2",
        "date_status_v2"
    ]
]

display(validation_df)


# ------------------------------------------------------------
# 10. RANGE VALIDATION
# ------------------------------------------------------------

valid_dates = shortlist_df[
    "parsed_interview_date_v2"
].dropna()

print("\n========== INTERVIEW DATE RANGE ==========")

if len(valid_dates) > 0:

    print(
        "Earliest interview date:",
        valid_dates.min().strftime("%Y-%m-%d")
    )

    print(
        "Latest interview date:",
        valid_dates.max().strftime("%Y-%m-%d")
    )

else:
    print("No valid interview dates found.")


# ------------------------------------------------------------
# 11. PRESERVATION CHECK
# ------------------------------------------------------------

print("\n========== VALIDATION ==========")

print(
    "Original shortlist records:",
    len(shortlist_df)
)

print(
    "Records preserved:",
    len(shortlist_df) == total_records
)

print(
    "Original 'Interview Date' column remains untouched:",
    "Interview Date" in shortlist_df.columns
)

print("\n" + "=" * 60)
print("STEP 11.2 COMPLETE")
print("=" * 60)
```

# Corrected Candidate + Role + Date Evidence Matching

```python
# ============================================================
# STEP 11.4 — CORRECTED CANDIDATE + ROLE + DATE MATCHING
# ============================================================

print("=" * 60)
print("STEP 11.4 — CORRECTED CANDIDATE + ROLE + DATE MATCHING")
print("=" * 60)

import pandas as pd
import numpy as np
import re
from difflib import SequenceMatcher


# ------------------------------------------------------------
# 1. Identify the master dataframe
# ------------------------------------------------------------

print("\n========== AVAILABLE DATAFRAMES ==========")

# FIXED: the original block printed evidence_match_df directly,
# which crashed with NameError when that variable wasn't in
# memory. This cell doesn't actually depend on reconciliation_df
# or evidence_match_df at all — everything it needs comes from
# candidate_df and shortlist_df below. This check now reports
# what's actually present instead of assuming and crashing.
for _name in ["reconciliation_df", "evidence_match_df", "candidate_df", "shortlist_df"]:
    if _name in globals():
        print(f"{_name:20}: {type(globals()[_name]).__name__}")
    else:
        print(f"{_name:20}: NOT FOUND")


# ------------------------------------------------------------
# 2. Helper functions
# ------------------------------------------------------------

def normalize_text(x):
    if pd.isna(x):
        return ""

    x = str(x).lower().strip()

    x = re.sub(r"[^a-z0-9\s]", " ", x)
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def name_similarity(a, b):

    a = normalize_text(a)
    b = normalize_text(b)

    if not a or not b:
        return 0.0

    if a == b:
        return 1.0

    # token-based comparison
    at = set(a.split())
    bt = set(b.split())

    if not at or not bt:
        return 0.0

    token_score = len(at & bt) / len(at | bt)

    sequence_score = SequenceMatcher(
        None,
        a,
        b
    ).ratio()

    return round(
        max(token_score, sequence_score),
        3
    )


def role_similarity(a, b):

    a = normalize_text(a)
    b = normalize_text(b)

    if not a or not b:
        return 0.0

    if a == b:
        return 1.0

    # Whole-word (token) membership, AND full-concept matching.
    #
    # FIX 1 (substring bug): use token sets, not "word in string",
    # so a short keyword like "ai" can't match by accident inside
    # an unrelated word (e.g. "ai" inside "complAInce").
    #
    # FIX 2 (split-keyword bug): a multi-word group like
    # ["sales", "engineer"] represents ONE compound role concept
    # ("sales engineer"). It must NOT award partial credit when
    # one side only contributes "sales" (e.g. "Area Sales Manager")
    # and the other side only contributes "engineer" (e.g.
    # "Research Testing Engineer") — those are unrelated roles
    # that each happen to contain one half of the phrase. A group
    # only counts when EVERY word in it is present on BOTH sides.
    a_tokens = set(a.split())
    b_tokens = set(b.split())

    role_groups = [
        ["software", "engineer"],
        ["full", "stack"],
        ["react"],
        ["node", "js"],
        ["agentic", "ai"],
        ["ai"],
        ["data", "analyst"],
        ["data", "analytics"],
        ["marketing", "data"],
        ["program", "manager"],
        ["project", "manager"],
        ["compliance"],
        ["sales", "engineer"],
        ["sales", "manager"],
        ["structural", "design"],
        ["content"],
        ["mep"],
        ["technical", "delivery"],
        ["product", "manager"],
        ["pmo"]
    ]

    best_group_score = 0.0

    for group in role_groups:

        a_hits = sum(
            word in a_tokens
            for word in group
        )

        b_hits = sum(
            word in b_tokens
            for word in group
        )

        # Require the FULL concept present on both sides —
        # no partial/averaged credit for split matches.
        if a_hits == len(group) and b_hits == len(group):

            best_group_score = max(
                best_group_score,
                1.0
            )

    seq_score = SequenceMatcher(
        None,
        a,
        b
    ).ratio()

    return round(
        max(best_group_score, seq_score * 0.70),
        3
    )


# ------------------------------------------------------------
# 3. Detect important columns automatically
# ------------------------------------------------------------

def find_column(df, possible_names):

    normalized = {
        normalize_text(c): c
        for c in df.columns
    }

    for name in possible_names:

        key = normalize_text(name)

        if key in normalized:
            return normalized[key]

    return None


short_name_col = find_column(
    shortlist_df,
    [
        "Candidate Name",
        "candidate_name",
        "Name"
    ]
)

short_role_col = find_column(
    shortlist_df,
    [
        "Position Applied",
        "shortlist_role",
        "Role"
    ]
)

short_recruiter_col = find_column(
    shortlist_df,
    [
        "Recuriter Name",
        "Recruiter Name",
        "recruiter"
    ]
)

print("\n========== SHORTLIST COLUMNS ==========")

print("Candidate:", short_name_col)
print("Role     :", short_role_col)
print("Recruiter:", short_recruiter_col)


# ------------------------------------------------------------
# 4. Master dataframe is the true 814-record candidate master,
#    not a shortlist-derived table. reconciliation_df and
#    evidence_match_df both only contain the ~59 shortlisted
#    candidates' OWN names (copied straight from shortlist_df
#    in earlier steps) — matching against either of them would
#    just compare the shortlist to itself instead of searching
#    the real master pool for genuine new matches.
# ------------------------------------------------------------

master_df = candidate_df.copy()

master_name_col = find_column(
    master_df,
    [
        "candidate_name",
        "Candidate Name",
        "Name"
    ]
)

master_role_col = find_column(
    master_df,
    [
        "position_applied",
        "Position Applied",
        "Role"
    ]
)

master_date_col = find_column(
    master_df,
    [
        "date",
        "_master_date",
        "Interview Date"
    ]
)

print("\n========== MASTER COLUMNS ==========")

print("Candidate:", master_name_col)
print("Role     :", master_role_col)
print("Date     :", master_date_col)

if master_name_col is None:
    raise KeyError(
        "Could not identify the candidate-name column in candidate_df."
    )


# ------------------------------------------------------------
# 5. Build shortlist working dataframe
# ------------------------------------------------------------

work = shortlist_df.copy()

work["_shortlist_index"] = range(len(work))

work["_candidate_name"] = (
    work[short_name_col]
    .astype(str)
)

work["_shortlist_role"] = (
    work[short_role_col]
    .astype(str)
)


# ------------------------------------------------------------
# 6. Get master candidate names
# ------------------------------------------------------------

master_df["_master_name"] = (
    master_df[master_name_col]
    .astype(str)
)


# ------------------------------------------------------------
# 7. Candidate-first matching
# ------------------------------------------------------------

results = []

for idx, row in work.iterrows():

    shortlist_name = row["_candidate_name"]
    shortlist_role = row["_shortlist_role"]

    shortlist_recruiter = (
        row[short_recruiter_col]
        if short_recruiter_col
        else None
    )

    best_candidates = []

    for m_idx, mrow in master_df.iterrows():

        master_name = mrow["_master_name"]

        name_score = name_similarity(
            shortlist_name,
            master_name
        )

        # Candidate identity gets priority.
        if name_score >= 0.85:

            best_candidates.append(
                (
                    m_idx,
                    name_score
                )
            )

    # --------------------------------------------------------
    # No candidate-level match
    # --------------------------------------------------------

    if len(best_candidates) == 0:

        results.append({
            "shortlist_index": idx,
            "master_row_index": np.nan,
            "candidate_name": shortlist_name,
            "shortlist_role": shortlist_role,
            "recruiter": shortlist_recruiter,
            "master_candidate_name": None,
            "master_role": None,
            "name_score_v2": 0.0,
            "role_score_v2": 0.0,
            "candidate_match_count_v2": 0,
            "candidate_match_status_v2": "UNMATCHED"
        })

        continue


    # --------------------------------------------------------
    # Evaluate role ONLY among candidate-name matches
    # --------------------------------------------------------

    candidate_rows = []

    for m_idx, name_score in best_candidates:

        mrow = master_df.loc[m_idx]

        master_role = (
            mrow[master_role_col]
            if master_role_col is not None
            else ""
        )

        r_score = role_similarity(
            shortlist_role,
            master_role
        )

        candidate_rows.append(
            (
                m_idx,
                name_score,
                r_score
            )
        )


    # Sort by name first, then role
    candidate_rows = sorted(
        candidate_rows,
        key=lambda x: (
            x[1],
            x[2]
        ),
        reverse=True
    )

    best_idx, best_name_score, best_role_score = (
        candidate_rows[0]
    )

    best_master = master_df.loc[best_idx]

    best_master_role = (
        best_master[master_role_col]
        if master_role_col is not None
        else None
    )

    # --------------------------------------------------------
    # Duplicate candidate names
    # --------------------------------------------------------

    if len(candidate_rows) > 1:

        top_name = candidate_rows[0][1]
        second_name = candidate_rows[1][1]

        if (
            abs(top_name - second_name) < 0.05
        ):

            candidate_status = (
                "DUPLICATE NAME - REVIEW"
            )

        else:

            candidate_status = (
                "NAME MATCH"
            )

    else:

        candidate_status = "NAME MATCH"


    results.append({
        "shortlist_index": idx,
        "master_row_index": best_idx,
        "candidate_name": shortlist_name,
        "shortlist_role": shortlist_role,
        "recruiter": shortlist_recruiter,
        "master_candidate_name": best_master["_master_name"],
        "master_role": best_master_role,
        "name_score_v2": best_name_score,
        "role_score_v2": best_role_score,
        "candidate_match_count_v2": len(candidate_rows),
        "candidate_match_status_v2": candidate_status
    })


# ------------------------------------------------------------
# 8. Create candidate-first reconciliation
# ------------------------------------------------------------

candidate_recon_v2 = pd.DataFrame(results)


# ------------------------------------------------------------
# 9. Add robust interview dates from Step 11.2
# ------------------------------------------------------------

if "parsed_interview_date_v2" in shortlist_df.columns:

    candidate_recon_v2[
        "parsed_interview_date_v2"
    ] = candidate_recon_v2[
        "shortlist_index"
    ].map(
        shortlist_df[
            "parsed_interview_date_v2"
        ]
    )

elif date_col if "date_col" in locals() else False:

    candidate_recon_v2[
        "parsed_interview_date_v2"
    ] = candidate_recon_v2[
        "shortlist_index"
    ].map(
        shortlist_df[date_col]
    )

else:

    candidate_recon_v2[
        "parsed_interview_date_v2"
    ] = pd.NaT


# ------------------------------------------------------------
# 10. Master date
# ------------------------------------------------------------

if master_date_col is not None:

    candidate_recon_v2[
        "master_date_v2"
    ] = candidate_recon_v2[
        "master_row_index"
    ].apply(
        lambda x:
        master_df.loc[x, master_date_col]
        if pd.notna(x) and x in master_df.index
        else pd.NaT
    )

    candidate_recon_v2[
        "master_date_v2"
    ] = pd.to_datetime(
        candidate_recon_v2[
            "master_date_v2"
        ],
        errors="coerce"
    )

else:

    candidate_recon_v2[
        "master_date_v2"
    ] = pd.NaT


# ------------------------------------------------------------
# 11. Date gap
# ------------------------------------------------------------

candidate_recon_v2[
    "date_gap_days_v2"
] = (
    pd.to_datetime(
        candidate_recon_v2[
            "parsed_interview_date_v2"
        ],
        errors="coerce"
    )
    -
    candidate_recon_v2[
        "master_date_v2"
    ]
).abs().dt.days


# ------------------------------------------------------------
# 12. Date evidence
# ------------------------------------------------------------

def date_score_v2(gap):

    if pd.isna(gap):
        return 0.0

    if gap == 0:
        return 1.0

    if gap <= 1:
        return 0.90

    if gap <= 3:
        return 0.75

    if gap <= 7:
        return 0.50

    if gap <= 14:
        return 0.25

    return 0.0


candidate_recon_v2[
    "date_score_v2"
] = candidate_recon_v2[
    "date_gap_days_v2"
].apply(
    date_score_v2
)


# ------------------------------------------------------------
# 13. Final evidence score
# ------------------------------------------------------------

candidate_recon_v2[
    "evidence_score_v2"
] = (
    0.50 *
    candidate_recon_v2[
        "name_score_v2"
    ]
    +
    0.30 *
    candidate_recon_v2[
        "role_score_v2"
    ]
    +
    0.20 *
    candidate_recon_v2[
        "date_score_v2"
    ]
).round(3)


# ------------------------------------------------------------
# 14. Final classification
# ------------------------------------------------------------

def classify_v2(row):

    name_score = row["name_score_v2"]
    role_score = row["role_score_v2"]
    date_score = row["date_score_v2"]
    evidence = row["evidence_score_v2"]

    if name_score < 0.85:

        return "UNMATCHED"

    if (
        row["candidate_match_status_v2"]
        == "DUPLICATE NAME - REVIEW"
    ):

        return "DUPLICATE NAME - REVIEW"

    if (
        name_score >= 0.95
        and role_score >= 0.65
        and evidence >= 0.75
    ):

        return "HIGH CONFIDENCE"

    if (
        name_score >= 0.90
        and role_score >= 0.40
    ):

        return "NAME + ROLE SUPPORT"

    return "NAME MATCH - ROLE REVIEW"


candidate_recon_v2[
    "decision_v2"
] = candidate_recon_v2.apply(
    classify_v2,
    axis=1
)


# ------------------------------------------------------------
# 15. Confidence
# ------------------------------------------------------------

def confidence_v2(score):

    if score >= 0.75:
        return "HIGH"

    if score >= 0.45:
        return "MEDIUM"

    return "LOW"


candidate_recon_v2[
    "confidence_v2"
] = candidate_recon_v2[
    "evidence_score_v2"
].apply(
    confidence_v2
)


# ------------------------------------------------------------
# 16. Summary
# ------------------------------------------------------------

print("\n========== CANDIDATE-FIRST MATCH SUMMARY ==========")

print(
    candidate_recon_v2[
        "decision_v2"
    ].value_counts()
)


print("\n========== NAME MATCH SUMMARY ==========")

print(
    candidate_recon_v2[
        "candidate_match_status_v2"
    ].value_counts()
)


print("\n========== CONFIDENCE SUMMARY ==========")

print(
    candidate_recon_v2[
        "confidence_v2"
    ].value_counts()
)


print("\n========== EVIDENCE SCORE SUMMARY ==========")

print(
    candidate_recon_v2[
        "evidence_score_v2"
    ].describe()
)


# ------------------------------------------------------------
# 17. Validation
# ------------------------------------------------------------

print("\n========== VALIDATION ==========")

print(
    "Original shortlist records :",
    len(shortlist_df)
)

print(
    "Reconciliation records     :",
    len(candidate_recon_v2)
)

print(
    "Records preserved          :",
    len(shortlist_df)
    == len(candidate_recon_v2)
)


# ------------------------------------------------------------
# 18. Display report
# ------------------------------------------------------------

display_cols = [
    "shortlist_index",
    "candidate_name",
    "shortlist_role",
    "recruiter",
    "master_candidate_name",
    "master_role",
    "name_score_v2",
    "role_score_v2",
    "master_date_v2",
    "parsed_interview_date_v2",
    "date_gap_days_v2",
    "date_score_v2",
    "evidence_score_v2",
    "decision_v2",
    "confidence_v2"
]

display_cols = [
    c for c in display_cols
    if c in candidate_recon_v2.columns
]

print("\n========== STEP 11.4 REPORT ==========")

display(
    candidate_recon_v2[
        display_cols
    ]
)


# ------------------------------------------------------------
# 19. Save final Step 11.4 dataframe under the exact name
#     Step 11.5 checks FIRST — otherwise Step 11.5 silently
#     falls back to the older, simpler Step 11 dataframe.
# ------------------------------------------------------------

evidence_match_df_v2 = candidate_recon_v2.copy()
reconciliation_v2 = candidate_recon_v2.copy()  # kept for compatibility

print("\n" + "=" * 60)
print("STEP 11.4 COMPLETE")
print("=" * 60)
```

# FINAL RECONCILIATION DATASET

```python
# ============================================================
# STEP 11.5 — FINAL EVIDENCE-BASED DECISION RULES
# ============================================================
#
# REVIEW tier role-score gate set to 0.35 (adjusted from an
# initial 0.65). Sensitivity testing in Step 11.5B showed the
# 0.65 gate was dumping confirmed name matches with reasonable
# overall evidence into LOW EVIDENCE - REVIEW purely because
# master role text is often generic/inconsistent in the source
# data (e.g. many roles logged simply as "Prowess - Compliance
# Manager" regardless of actual position). 0.35 lets evidence
# quality (which already weighs name/role/date together) decide
# REVIEW vs LOW EVIDENCE, rather than role text alone gatekeeping
# it. HIGH CONFIDENCE's 0.85 role bar is untouched — this only
# reclassifies the weaker tier, it does not inflate top-tier
# matches.
# ============================================================

import pandas as pd
import numpy as np

print("=" * 60)
print("STEP 11.5 — FINAL EVIDENCE-BASED DECISION RULES")
print("=" * 60)

# ------------------------------------------------------------
# 1. Identify the Step 11.4 dataframe
# ------------------------------------------------------------

if "evidence_match_df_v2" in globals():
    final_df = evidence_match_df_v2.copy()

elif "evidence_match_df" in globals():
    final_df = evidence_match_df.copy()

elif "reconciliation_df_v2" in globals():
    final_df = reconciliation_df_v2.copy()

elif "reconciliation_df" in globals():
    final_df = reconciliation_df.copy()

else:
    raise NameError(
        "Step 11.4 dataframe not found. "
        "Expected evidence_match_df_v2, evidence_match_df, "
        "reconciliation_df_v2 or reconciliation_df."
    )

print(f"Input records from Step 11.4 : {len(final_df)}")


# ------------------------------------------------------------
# 2. Helper — safely find columns
# ------------------------------------------------------------

def find_col(df, possible_names):

    for col in possible_names:
        if col in df.columns:
            return col

    return None


candidate_col = find_col(
    final_df,
    [
        "candidate_name",
        "master_candidate_name",
        "Candidate Name",
        "candidate",
        "shortlist_name"
    ]
)

shortlist_name_col = find_col(
    final_df,
    [
        "shortlist_name",
        "candidate_name",
        "Candidate Name"
    ]
)

role_score_col = find_col(
    final_df,
    [
        "role_score_v2",
        "role_score",
        "Role Score"
    ]
)

date_gap_col = find_col(
    final_df,
    [
        "date_gap_days_v2",
        "date_gap_days",
        "date_gap"
    ]
)

date_score_col = find_col(
    final_df,
    [
        "date_score_v2",
        "date_score"
    ]
)

evidence_score_col = find_col(
    final_df,
    [
        "evidence_score_v2",
        "evidence_score"
    ])


print("\n========== DETECTED COLUMNS ==========")
print("Candidate column :", candidate_col)
print("Role score       :", role_score_col)
print("Date gap         :", date_gap_col)
print("Date score       :", date_score_col)
print("Evidence score   :", evidence_score_col)


# ------------------------------------------------------------
# 3. Convert scoring columns safely
# ------------------------------------------------------------

for col in [
    role_score_col,
    date_gap_col,
    date_score_col,
    evidence_score_col
]:
    if col is not None:
        final_df[col] = pd.to_numeric(
            final_df[col],
            errors="coerce"
        )


# ------------------------------------------------------------
# 4. Create normalized working columns
# ------------------------------------------------------------

final_df["_role_score"] = (
    final_df[role_score_col]
    if role_score_col
    else 0
)

final_df["_date_gap"] = (
    final_df[date_gap_col]
    if date_gap_col
    else np.nan
)

final_df["_date_score"] = (
    final_df[date_score_col]
    if date_score_col
    else 0
)

final_df["_evidence_score"] = (
    final_df[evidence_score_col]
    if evidence_score_col
    else 0
)


# ------------------------------------------------------------
# 5. Final decision rules
#
# IMPORTANT:
# Name matching has already happened in previous steps.
# Here we only classify the evidence quality.
#
# HIGH CONFIDENCE:
#   Strong role evidence + close interview date
#
# REVIEW:
#   Reasonable evidence but not strong enough
#   (role gate: 0.35 — see note at top of cell)
#
# LOW EVIDENCE - REVIEW:
#   Weak role/date evidence
#
# UNMATCHED:
#   No usable master evidence
#
# DUPLICATE - REVIEW:
#   Duplicate master-name situation
# ------------------------------------------------------------

def final_decision(row):

    role = row["_role_score"]
    gap = row["_date_gap"]
    date_score = row["_date_score"]
    evidence = row["_evidence_score"]

    # --------------------------------------------------------
    # No evidence
    # --------------------------------------------------------
    if pd.isna(role) or role <= 0:
        return "UNMATCHED"

    # --------------------------------------------------------
    # Duplicate flag from previous steps
    # --------------------------------------------------------
    existing_decision = str(
        row.get("decision_v2", "")
    ).upper()

    existing_classification = str(
        row.get("decision", "")
    ).upper()

    if (
        "DUPLICATE" in existing_decision
        or "DUPLICATE" in existing_classification
    ):
        return "DUPLICATE - REVIEW"

    # --------------------------------------------------------
    # HIGH CONFIDENCE
    #
    # Strong role match AND date evidence.
    #
    # We intentionally restrict date gap.
    # --------------------------------------------------------

    if (
        role >= 0.85
        and
        not pd.isna(gap)
        and
        gap <= 14
        and
        evidence >= 0.60
    ):
        return "HIGH CONFIDENCE"

    # --------------------------------------------------------
    # REVIEW
    #
    # Good role evidence but date evidence is weaker,
    # or moderate overall evidence.
    #
    # Role gate lowered to 0.35 (was 0.65) — see cell header.
    # --------------------------------------------------------

    if (
        role >= 0.35
        and
        (
            pd.isna(gap)
            or gap <= 30
        )
        and
        evidence >= 0.40
    ):
        return "REVIEW"

    # --------------------------------------------------------
    # REVIEW for strong role but larger date gap
    # --------------------------------------------------------

    if (
        role >= 0.85
        and
        not pd.isna(gap)
        and
        gap > 14
    ):
        return "REVIEW"

    # --------------------------------------------------------
    # LOW EVIDENCE
    # --------------------------------------------------------

    return "LOW EVIDENCE - REVIEW"


# ------------------------------------------------------------
# 6. Apply final decision
# ------------------------------------------------------------

final_df["final_decision"] = final_df.apply(
    final_decision,
    axis=1
)


# ------------------------------------------------------------
# 7. Final confidence
# ------------------------------------------------------------

def final_confidence(decision):

    if decision == "HIGH CONFIDENCE":
        return "HIGH"

    elif decision == "REVIEW":
        return "MEDIUM"

    elif decision == "DUPLICATE - REVIEW":
        return "LOW"

    else:
        return "LOW"


final_df["final_confidence"] = (
    final_df["final_decision"]
    .apply(final_confidence)
)


# ------------------------------------------------------------
# 8. Human-readable reason
# ------------------------------------------------------------

def build_reason(row):

    decision = row["final_decision"]
    role = row["_role_score"]
    gap = row["_date_gap"]
    evidence = row["_evidence_score"]

    if decision == "UNMATCHED":
        return "No usable master candidate evidence"

    if decision == "DUPLICATE - REVIEW":
        return "Duplicate candidate name requires manual review"

    if decision == "HIGH CONFIDENCE":
        return (
            f"Strong role match ({role:.2f}) and "
            f"interview date within {gap:.0f} days"
        )

    if decision == "REVIEW":
        if not pd.isna(gap):
            return (
                f"Potential role match ({role:.2f}) "
                f"with {gap:.0f}-day date gap"
            )
        return (
            f"Potential role match ({role:.2f}) "
            "but date evidence is incomplete"
        )

    return (
        f"Insufficient evidence "
        f"(role={role:.2f}, evidence={evidence:.2f})"
    )


final_df["final_reason"] = final_df.apply(
    build_reason,
    axis=1
)


# ------------------------------------------------------------
# 9. FINAL SUMMARY
# ------------------------------------------------------------

print("\n========== FINAL DECISION SUMMARY ==========")

print(
    final_df["final_decision"]
    .value_counts()
    .to_string()
)


print("\n========== FINAL CONFIDENCE SUMMARY ==========")

print(
    final_df["final_confidence"]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 10. Evidence statistics
# ------------------------------------------------------------

print("\n========== FINAL EVIDENCE SUMMARY ==========")

print(
    final_df["_evidence_score"]
    .describe()
)


# ------------------------------------------------------------
# 11. Final reconciliation report
# ------------------------------------------------------------

display_cols = [
    col for col in [
        "shortlist_index",
        "candidate_name",
        "shortlist_name",
        "master_candidate_name",
        "shortlist_role",
        "master_role",
        "role_score",
        "role_score_v2",
        "master_date",
        "parsed_interview_date_v2",
        "date_gap_days_v2",
        "date_score_v2",
        "evidence_score_v2",
        "final_decision",
        "final_confidence",
        "final_reason"
    ]
    if col in final_df.columns
]

print("\n========== FINAL RECONCILIATION REPORT ==========")

display(
    final_df[display_cols]
)


# ------------------------------------------------------------
# 12. HIGH CONFIDENCE ONLY
# ------------------------------------------------------------

high_confidence_df = final_df[
    final_df["final_decision"] == "HIGH CONFIDENCE"
].copy()

print("\n========== HIGH CONFIDENCE MATCHES ==========")

print(
    f"High-confidence records : {len(high_confidence_df)}"
)

if len(high_confidence_df) > 0:
    display(
        high_confidence_df[display_cols]
    )
else:
    print("No high-confidence records found.")


# ------------------------------------------------------------
# 13. REVIEW RECORDS
# ------------------------------------------------------------

review_df = final_df[
    final_df["final_decision"].isin([
        "REVIEW",
        "LOW EVIDENCE - REVIEW",
        "DUPLICATE - REVIEW"
    ])
].copy()

print("\n========== RECORDS REQUIRING REVIEW ==========")

print(
    f"Review records : {len(review_df)}"
)

if len(review_df) > 0:
    display(
        review_df[display_cols]
    )


# ------------------------------------------------------------
# 14. UNMATCHED
# ------------------------------------------------------------

unmatched_df = final_df[
    final_df["final_decision"] == "UNMATCHED"
].copy()

print("\n========== UNMATCHED RECORDS ==========")

print(
    f"Unmatched records : {len(unmatched_df)}"
)

if len(unmatched_df) > 0:
    display(
        unmatched_df[display_cols]
    )


# ------------------------------------------------------------
# 15. VALIDATION
# ------------------------------------------------------------

print("\n========== VALIDATION ==========")

print(
    "Input records       :",
    len(final_df)
)

print(
    "Output records      :",
    len(final_df)
)

print(
    "Records preserved   :",
    len(final_df) == 59
)

print(
    "Decision assigned   :",
    final_df["final_decision"].notna().all()
)

print(
    "Confidence assigned :",
    final_df["final_confidence"].notna().all()
)

print(
    "Original records preserved:",
    len(final_df) == len(
        final_df.drop_duplicates(
            subset=["shortlist_index"]
        )
    )
)


print("\n" + "=" * 60)
print("STEP 11.5 COMPLETE")
print("=" * 60)
```

# CLIENT-WISE RECRUITMENT PERFORMANCE ANALYSIS

```python
# ============================================================
# STEP 20 — CLIENT-WISE RECRUITMENT PERFORMANCE ANALYSIS
# FIXED VERSION
# ============================================================

print("=" * 60)
print("STEP 20 — CLIENT-WISE RECRUITMENT PERFORMANCE ANALYSIS")
print("=" * 60)

# ------------------------------------------------------------
# INPUT
# ------------------------------------------------------------

input_df = business_funnel_analysis_df.copy()

print("\n========== INPUT VALIDATION ==========")
print("Input records :", len(input_df))
print("Input columns :", len(input_df.columns))

# ------------------------------------------------------------
# CHECK AVAILABLE COLUMNS
# ------------------------------------------------------------

print("\nAvailable columns:")
print(input_df.columns.tolist())

# ------------------------------------------------------------
# CREATE SELECTION OUTCOME IF NOT PRESENT
# ------------------------------------------------------------

if "selection_outcome" not in input_df.columns:

    if "selection_status" not in input_df.columns:
        raise ValueError(
            "Neither selection_outcome nor selection_status "
            "is available in the dataset."
        )

    def classify_selection(value):

        if pd.isna(value):
            return "No Selection Status"

        text = str(value).strip().lower()

        if text in ["", "missing", "nan", "none"]:
            return "No Selection Status"

        # Check rejection BEFORE shortlisted — "shortlisted" is
        # a substring of "not shortlisted", so checking for
        # "shortlisted" first would classify every rejection
        # ("not shortlisted") as a Shortlisted record. Checking
        # the more specific rejection pattern first avoids this.
        if (
            "not shortlisted" in text
            or "not selected" in text
            or "reject" in text
        ):
            return "Not Shortlisted / Rejected"

        if "shortlisted" in text:
            return "Shortlisted"

        if "resume shared" in text or "resume sent" in text:
            return "Resume / Profile Shared"

        return "Other / Interview Activity"

    input_df["selection_outcome"] = (
        input_df["selection_status"]
        .apply(classify_selection)
    )

    print("\nSelection outcome recreated from selection_status : True")

else:
    print("\nSelection outcome already available : True")


# ------------------------------------------------------------
# REQUIRED COLUMN CHECK
# ------------------------------------------------------------

required_columns = [
    "client_business",
    "candidate_name",
    "status_category",
    "rm_pipeline_category",
    "selection_outcome",
    "position_applied"
]

missing_columns = [
    col for col in required_columns
    if col not in input_df.columns
]

print("\nMissing required columns :", missing_columns)

if missing_columns:
    raise ValueError(f"Missing required columns: {missing_columns}")


# ------------------------------------------------------------
# CLIENT-WISE RECRUITMENT VOLUME
# ------------------------------------------------------------

client_volume = (
    input_df
    .groupby("client_business")
    .size()
    .reset_index(name="candidate_count")
)

client_volume["percentage"] = (
    client_volume["candidate_count"]
    / len(input_df) * 100
).round(2)

client_volume = client_volume.sort_values(
    "candidate_count",
    ascending=False
).reset_index(drop=True)

print("\n========== CLIENT-WISE RECRUITMENT VOLUME ==========")
print(client_volume.to_string(index=False))


# ------------------------------------------------------------
# CLIENT-WISE PIPELINE COVERAGE
# ------------------------------------------------------------

client_status = (
    input_df
    .groupby("client_business")
    .agg(
        total_candidates=("candidate_name", "count"),

        records_with_status=(
            "status_category",
            lambda x: (x != "NO STATUS").sum()
        ),

        records_with_rm_status=(
            "rm_pipeline_category",
            lambda x: (x != "NO RM STATUS").sum()
        ),

        records_with_selection=(
            "selection_outcome",
            lambda x: (x != "No Selection Status").sum()
        )
    )
    .reset_index()
)

client_status["status_coverage_pct"] = (
    client_status["records_with_status"]
    / client_status["total_candidates"] * 100
).round(2)

client_status["rm_coverage_pct"] = (
    client_status["records_with_rm_status"]
    / client_status["total_candidates"] * 100
).round(2)

client_status["selection_coverage_pct"] = (
    client_status["records_with_selection"]
    / client_status["total_candidates"] * 100
).round(2)

print("\n========== CLIENT-WISE PIPELINE COVERAGE ==========")
print(client_status.to_string(index=False))


# ------------------------------------------------------------
# CLIENT-WISE SELECTION OUTCOME
# ------------------------------------------------------------

client_selection = (
    input_df
    .groupby("client_business")
    .agg(
        total_candidates=("candidate_name", "count"),

        shortlisted=(
            "selection_outcome",
            lambda x: (x == "Shortlisted").sum()
        ),

        profile_shared=(
            "selection_outcome",
            lambda x: (x == "Resume / Profile Shared").sum()
        ),

        rejected=(
            "selection_outcome",
            lambda x: (x == "Not Shortlisted / Rejected").sum()
        ),

        # Anything with a selection status that isn't one of the
        # three named categories above (e.g. "Other / Interview
        # Activity" or an uncategorized value) — makes this
        # breakdown exhaustive instead of silently dropping
        # records, which previously left client-level totals
        # that didn't reconcile against records_with_selection.
        other_selection_activity=(
            "selection_outcome",
            lambda x: (
                ~x.isin([
                    "No Selection Status",
                    "Shortlisted",
                    "Resume / Profile Shared",
                    "Not Shortlisted / Rejected"
                ])
            ).sum()
        )
    )
    .reset_index()
)

client_selection["shortlist_rate_pct"] = (
    client_selection["shortlisted"]
    / client_selection["total_candidates"] * 100
).round(2)

print("\n========== CLIENT-WISE SELECTION OUTCOME ==========")
print(client_selection.to_string(index=False))


# ------------------------------------------------------------
# CLIENT-WISE TOP ROLE
# ------------------------------------------------------------

role_counts = (
    input_df
    .groupby(
        ["client_business", "position_applied"]
    )
    .size()
    .reset_index(name="candidate_count")
)

role_counts = role_counts.sort_values(
    ["client_business", "candidate_count"],
    ascending=[True, False]
)

top_role_by_client = (
    role_counts
    .groupby("client_business", as_index=False)
    .first()
)

top_role_by_client = top_role_by_client.rename(
    columns={
        "position_applied": "top_role",
        "candidate_count": "top_role_candidate_count"
    }
)

print("\n========== TOP ROLE BY CLIENT ==========")
print(top_role_by_client.to_string(index=False))


# ------------------------------------------------------------
# FINAL CLIENT PERFORMANCE DATAFRAME
# ------------------------------------------------------------

client_performance_df = (
    client_volume
    .merge(
        client_status,
        on="client_business",
        how="left"
    )
    .merge(
        client_selection[
            [
                "client_business",
                "shortlisted",
                "profile_shared",
                "rejected",
                "other_selection_activity",
                "shortlist_rate_pct"
            ]
        ],
        on="client_business",
        how="left"
    )
    .merge(
        top_role_by_client[
            [
                "client_business",
                "top_role",
                "top_role_candidate_count"
            ]
        ],
        on="client_business",
        how="left"
    )
)


# ------------------------------------------------------------
# FINAL OUTPUT
# ------------------------------------------------------------

print("\n========== CLIENT PERFORMANCE DATAFRAME ==========")
print("Dataframe :", "client_performance_df")
print("Records   :", len(client_performance_df))
print("Columns   :", len(client_performance_df.columns))

print("\n========== CLIENT PERFORMANCE ==========")
print(client_performance_df.to_string(index=False))


# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

print("\n========== STEP 20 VALIDATION ==========")

print(
    "Client groups :",
    client_performance_df["client_business"].nunique()
)

print(
    "Total records represented :",
    client_performance_df["total_candidates"].sum()
)

print(
    "Original 814 records preserved :",
    client_performance_df["total_candidates"].sum() == len(input_df)
)

print(
    "Business names available :",
    client_performance_df["client_business"].notna().all()
)

print(
    "59-record reconciliation preserved :",
    len(final_reconciliation_df) == 59
)

# NEW: cross-check that selection breakdown reconciles with
# the coverage table's records_with_selection count.
selection_reconciles = (
    (
        client_performance_df["shortlisted"]
        + client_performance_df["profile_shared"]
        + client_performance_df["rejected"]
        + client_performance_df["other_selection_activity"]
    )
    == client_performance_df["records_with_selection"]
).all()

print(
    "Selection breakdown reconciles per client :",
    selection_reconciles
)

print(
    "Performance dataframe created :",
    True
)

print("\n" + "=" * 60)
print("STEP 20 COMPLETE")
print("=" * 60)
```

# CANDIDATE 360° & REUSABLE TALENT POOL

```python
# ============================================================
# STEP 22 — CANDIDATE 360° & REUSABLE TALENT POOL
# ============================================================

import pandas as pd
import numpy as np
import re

print("=" * 60)
print("STEP 22 — CANDIDATE 360° & REUSABLE TALENT POOL")
print("=" * 60)

# ------------------------------------------------------------
# 1. INPUT VALIDATION
# ------------------------------------------------------------


if "confidential_business_analysis_df" in globals():
    source_df = confidential_business_analysis_df.copy()
elif "business_analysis_df" in globals():
    source_df = business_analysis_df.copy()
else:
    raise NameError(
        "Required 814-record dataframe not found. "
        "Expected confidential_business_analysis_df."
    )

print("\n========== INPUT VALIDATION ==========")
print("Input records :", len(source_df))
print("Input columns :", len(source_df.columns))

if len(source_df) != 814:
    raise ValueError(
        f"Expected 814 records, but found {len(source_df)}"
    )

required_cols = [
    "date",
    "candidate_name",
    "position_applied",
    "client_business",
    "status",
    "rm_status",
    "selection_status"
]

missing_cols = [c for c in required_cols if c not in source_df.columns]

if missing_cols:
    raise ValueError(f"Missing required columns: {missing_cols}")

print("Required columns available :", True)
print("814-record population intact:", True)


# ------------------------------------------------------------
# 2. CREATE CLEAN WORKING COPY
# ------------------------------------------------------------

work_22 = source_df.copy()

# Preserve original values
work_22["candidate_name_original"] = work_22["candidate_name"]

# Normalize candidate names only for grouping
work_22["candidate_key"] = (
    work_22["candidate_name"]
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

# Remove obvious null-like values
work_22["candidate_key"] = work_22["candidate_key"].replace(
    {
        "": np.nan,
        "nan": np.nan,
        "none": np.nan,
        "null": np.nan
    }
)

# Date
work_22["date_clean"] = pd.to_datetime(
    work_22["date"],
    errors="coerce"
)

# ------------------------------------------------------------
# 3. CANDIDATE 360° PROFILE
# ------------------------------------------------------------

def unique_join(series):
    values = []
    for x in series.dropna():
        x = str(x).strip()
        if x and x.lower() not in ["nan", "none", "missing"]:
            if x not in values:
                values.append(x)
    return " | ".join(values)


candidate_360_df = (
    work_22
    .dropna(subset=["candidate_key"])
    .sort_values(["candidate_key", "date_clean"])
    .groupby("candidate_key", as_index=False)
    .agg(
        candidate_name=("candidate_name_original", "first"),
        recruitment_records=("candidate_key", "size"),

        clients=("client_business", unique_join),
        roles=("position_applied", unique_join),
        recruitment_dates=("date_clean", unique_join),

        statuses=("status", unique_join),
        rm_history=("rm_status", unique_join),
        selection_history=("selection_status", unique_join)
    )
)

# ------------------------------------------------------------
# 4. CANDIDATE HISTORY COUNTS
# ------------------------------------------------------------

def count_unique_values(series):
    values = set()

    for x in series.dropna():
        text = str(x).strip()

        if not text:
            continue

        for item in text.split(" | "):
            item = item.strip()
            if item:
                values.add(item)

    return len(values)


candidate_360_df["client_count"] = candidate_360_df["clients"].apply(
    lambda x: len([v for v in str(x).split(" | ") if v.strip()])
)

candidate_360_df["role_count"] = candidate_360_df["roles"].apply(
    lambda x: len([v for v in str(x).split(" | ") if v.strip()])
)

candidate_360_df["recruitment_date_count"] = candidate_360_df[
    "recruitment_dates"
].apply(
    lambda x: len([v for v in str(x).split(" | ") if v.strip()])
)

# ------------------------------------------------------------
# 5. LATEST RECRUITMENT RECORD
# ------------------------------------------------------------

latest_records = (
    work_22
    .dropna(subset=["candidate_key"])
    .sort_values(["candidate_key", "date_clean"])
    .groupby("candidate_key", as_index=False)
    .tail(1)
    .copy()
)

latest_records = latest_records[
    [
        "candidate_key",
        "date_clean",
        "status",
        "rm_status",
        "selection_status",
        "position_applied",
        "client_business"
    ]
].rename(
    columns={
        "date_clean": "latest_recruitment_date",
        "status": "latest_status",
        "rm_status": "latest_rm_status",
        "selection_status": "latest_selection_status",
        "position_applied": "latest_role",
        "client_business": "latest_client_business"
    }
)

candidate_360_df = candidate_360_df.merge(
    latest_records,
    on="candidate_key",
    how="left"
)

# ------------------------------------------------------------
# 6. REUSABLE CANDIDATE LOGIC
# ------------------------------------------------------------

# Positive / potentially reusable recruitment signals.
positive_patterns = [
    "interested",
    "call back",
    "callback",
    "mail sent",
    "profile shared",
    "resume shared",
    "pipeline",
    "waiting",
    "interview",
    "shortlisted",
    "imm.joiner",
    "immediate joiner",
    "joined"
]

negative_patterns = [
    "not interested",
    "not responded",
    "not looking",
    "not ok",
    "rejected",
    "not shortlisted",
    "not selected",
    "denied"
]


def contains_any(text, patterns):
    text = str(text).lower()
    return any(p in text for p in patterns)


# Two negative patterns contain a positive pattern as a literal
# substring — "not interested" contains "interested", and
# "not shortlisted" contains "shortlisted". We strip out negative
# phrases before checking positive patterns, so an embedded
# positive substring inside a negative phrase can't trigger a
# false positive signal.
def has_genuine_positive_signal(text):
    text = str(text).lower()
    cleaned = text
    for neg in negative_patterns:
        cleaned = cleaned.replace(neg, "")
    return contains_any(cleaned, positive_patterns)


# Combine available recruitment history
candidate_360_df["history_text"] = (
    candidate_360_df["statuses"].fillna("").astype(str)
    + " | "
    + candidate_360_df["rm_history"].fillna("").astype(str)
    + " | "
    + candidate_360_df["selection_history"].fillna("").astype(str)
)

candidate_360_df["negative_history_signal"] = candidate_360_df[
    "history_text"
].apply(
    lambda x: contains_any(x, negative_patterns)
)

candidate_360_df["positive_history_signal"] = candidate_360_df[
    "history_text"
].apply(
    has_genuine_positive_signal
)

# Candidate has demonstrated history across multiple roles/clients
candidate_360_df["multi_role_history"] = (
    candidate_360_df["role_count"] > 1
)

candidate_360_df["multi_client_history"] = (
    candidate_360_df["client_count"] > 1
)

# Reusable candidate:
# - multiple role history OR multiple client history
# - OR positive recruitment signal
# - but exclude candidates whose history is purely negative
candidate_360_df["potentially_reusable"] = (
    (
        candidate_360_df["multi_role_history"]
        | candidate_360_df["multi_client_history"]
        | candidate_360_df["positive_history_signal"]
    )
    & ~(
        candidate_360_df["negative_history_signal"]
        & ~candidate_360_df["positive_history_signal"]
    )
)

# ------------------------------------------------------------
# 7. REUSABLE POOL
# ------------------------------------------------------------

reusable_talent_pool_df = candidate_360_df[
    candidate_360_df["potentially_reusable"]
].copy()

# Priority score for future matching
reusable_talent_pool_df["reuse_score"] = (
    reusable_talent_pool_df["multi_role_history"].astype(int) * 3
    + reusable_talent_pool_df["multi_client_history"].astype(int) * 3
    + reusable_talent_pool_df["positive_history_signal"].astype(int) * 2
)

reusable_talent_pool_df = reusable_talent_pool_df.sort_values(
    ["reuse_score", "recruitment_records", "role_count"],
    ascending=[False, False, False]
).reset_index(drop=True)

# ------------------------------------------------------------
# 8. REUSABLE POOL SUMMARY
# ------------------------------------------------------------

reuse_summary_df = pd.DataFrame({
    "metric": [
        "Total unique candidates",
        "Candidates with multiple roles",
        "Candidates with multiple clients",
        "Candidates with positive history signal",
        "Potentially reusable candidates",
        "Single-history candidates"
    ],
    "count": [
        len(candidate_360_df),
        candidate_360_df["multi_role_history"].sum(),
        candidate_360_df["multi_client_history"].sum(),
        candidate_360_df["positive_history_signal"].sum(),
        candidate_360_df["potentially_reusable"].sum(),
        (~candidate_360_df["potentially_reusable"]).sum()
    ]
})

# ------------------------------------------------------------
# 8.5. ANONYMIZE ROLE TEXT FOR DISPLAY
# ------------------------------------------------------------
# "roles" and "latest_role" carry real client names directly
# next to real candidate names. Uses word-boundary matching so
# a client name is caught anywhere in the string — start,
# middle, or end, with or without a separator — rather than
# only matching specific prefix/suffix patterns (which missed
# cases like "full stack prowess", a client name with no
# separator at the end of the string).

def confidential_role_name(role):
    role = str(role).strip()

    exact_replacements = {
        "AI Specialist - Fischer Group Bangalore": "AI Specialist",
        "Senior Role - Safety L&T": "Senior Safety Role",
        "L & T Structural Design Coordinator": "Structural Design Coordinator",
        "Leviat - Sales Engineer": "Sales Engineer",
        "Research Testing Engineer - Hilti India": "Research Testing Engineer",
        "Lakshmi Hospital - Pharmacy In-Charge": "Pharmacy In-Charge",
        "Pharmacy in charge": "Pharmacy In-Charge",
    }
    if role in exact_replacements:
        return exact_replacements[role]

    client_word_pattern = re.compile(
        r"\b(prowess|lodha( group)?|fischer( group)?|hilti|alumil|"
        r"leviat|l\s*&\s*t|lakshmi\s*hospital)\b",
        flags=re.IGNORECASE
    )

    role = client_word_pattern.sub("", role)
    role = re.sub(r"\s*-\s*-\s*", " - ", role)
    role = re.sub(r"^[\s\-]+|[\s\-]+$", "", role)
    role = re.sub(r"\s+", " ", role)

    return role if role else "Role Undisclosed"


candidate_360_df["roles"] = candidate_360_df["roles"].apply(
    lambda x: " | ".join(
        confidential_role_name(r) for r in str(x).split(" | ") if r.strip()
    )
)

candidate_360_df["latest_role"] = candidate_360_df["latest_role"].apply(
    confidential_role_name
)

# Propagate the anonymized values into reusable_talent_pool_df,
# since it was sliced from candidate_360_df before this masking
# was applied.
reusable_talent_pool_df["roles"] = candidate_360_df.set_index(
    "candidate_key"
).loc[reusable_talent_pool_df["candidate_key"], "roles"].values

reusable_talent_pool_df["latest_role"] = candidate_360_df.set_index(
    "candidate_key"
).loc[reusable_talent_pool_df["candidate_key"], "latest_role"].values

# ------------------------------------------------------------
# 9. OUTPUT
# ------------------------------------------------------------

print("\n========== CANDIDATE 360° SUMMARY ==========")
print("Unique candidates :", len(candidate_360_df))
print(
    "Candidates with multiple roles :",
    int(candidate_360_df["multi_role_history"].sum())
)
print(
    "Candidates with multiple clients :",
    int(candidate_360_df["multi_client_history"].sum())
)
print(
    "Candidates with positive history :",
    int(candidate_360_df["positive_history_signal"].sum())
)

print("\n========== REUSABLE TALENT POOL ==========")
print(
    "Potentially reusable candidates :",
    len(reusable_talent_pool_df)
)

print(
    "Reusable pool percentage :",
    round(
        len(reusable_talent_pool_df)
        / len(candidate_360_df) * 100,
        2
    )
)

print("\n========== TOP REUSABLE CANDIDATES ==========")

display_cols = [
    "candidate_name",
    "recruitment_records",
    "client_count",
    "role_count",
    "latest_recruitment_date",
    "latest_role",
    "latest_status",
    "reuse_score"
]

print(
    reusable_talent_pool_df[display_cols]
    .head(20)
    .to_string(index=False)
)

print("\n========== REUSE SUMMARY ==========")
print(reuse_summary_df.to_string(index=False))

# ------------------------------------------------------------
# 10. VALIDATION
# ------------------------------------------------------------

print("\n========== STEP 22 VALIDATION ==========")

print(
    "Original 814 records preserved :",
    len(source_df) == 814
)

print(
    "Candidate 360 dataframe created :",
    len(candidate_360_df) > 0
)

print(
    "Reusable talent pool created :",
    len(reusable_talent_pool_df) >= 0
)

print(
    "Candidate names available :",
    candidate_360_df["candidate_name"].notna().all()
)

client_name_pattern = (
    r"Prowess|Lodha|Fischer|L&T|Leviat|Hilti|Alumil|Lakshmi Hospital"
)

confidential_check_passed = not (
    candidate_360_df["clients"]
    .astype(str)
    .str.contains(client_name_pattern, case=False, regex=True)
    .any()
    or candidate_360_df["roles"]
    .astype(str)
    .str.contains(client_name_pattern, case=False, regex=True)
    .any()
    or candidate_360_df["latest_role"]
    .astype(str)
    .str.contains(client_name_pattern, case=False, regex=True)
    .any()
)

print(
    "Confidential client names used :",
    confidential_check_passed
)

print(
    "59-record reconciliation preserved :",
    (
        "final_reconciliation_df" in globals()
        and len(final_reconciliation_df) == 59
    )
)

# Remove temporary helper column from final Candidate 360 output
candidate_360_df = candidate_360_df.drop(
    columns=["history_text"],
    errors="ignore"
)

print("\nCandidate 360° records :", len(candidate_360_df))
print("Reusable pool records  :", len(reusable_talent_pool_df))

print("\n" + "=" * 60)
print("STEP 22 COMPLETE")
print("=" * 60)
```

# JD ↔ Candidate Matching.

```python
# ============================================================
# STEP 23 — JD ↔ CANDIDATE MATCHING
# ============================================================

import pandas as pd
import numpy as np
import re

print("=" * 60)
print("STEP 23 — JD ↔ CANDIDATE MATCHING")
print("=" * 60)

# ------------------------------------------------------------
# 1. INPUT VALIDATION
# ------------------------------------------------------------

if "confidential_business_analysis_df" in globals():
    source_23 = confidential_business_analysis_df.copy()
else:
    raise NameError(
        "confidential_business_analysis_df not found."
    )

print("\n========== INPUT VALIDATION ==========")
print("Input records :", len(source_23))
print("Input columns :", len(source_23.columns))

if len(source_23) != 814:
    raise ValueError(
        f"Expected 814 records, found {len(source_23)}"
    )

required_cols = [
    "candidate_name",
    "position_applied",
    "client_business",
    "years_of_exp",
    "current_location",
    "job_location",
    "cctc",
    "ectc",
    "notice_period",
    "status"
]

missing_cols = [
    c for c in required_cols
    if c not in source_23.columns
]

if missing_cols:
    raise ValueError(
        f"Missing required columns: {missing_cols}"
    )

print("Required columns available :", True)


# ------------------------------------------------------------
# 2. TEXT NORMALIZATION
# ------------------------------------------------------------

def clean_text(value):
    if pd.isna(value):
        return ""
    return re.sub(
        r"\s+",
        " ",
        str(value).strip().lower()
    )


def token_set(value):
    text = clean_text(value)
    return set(
        re.findall(r"[a-zA-Z0-9]+", text)
    )


# ------------------------------------------------------------
# 2.5. NORMALIZED CANDIDATE IDENTITY
# ------------------------------------------------------------
# FIXED: previously grouped by raw candidate_name (Section 7),
# which is inconsistent with Step 22's approach — Step 22 groups
# by a normalized key specifically to merge whitespace-corrupted
# duplicates (e.g. "\n\nSatyam Jaysawal" vs "Satyam Jaysawal").
# Using the same normalized key here keeps candidate identity
# consistent across both steps, so the reusable-pool join in
# Section 13 doesn't silently miss a match due to a whitespace
# mismatch.

source_23["candidate_key"] = (
    source_23["candidate_name"]
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)


# ------------------------------------------------------------
# 3. ROLE NORMALIZATION
# ------------------------------------------------------------

source_23["role_text"] = (
    source_23["position_applied"]
    .fillna("")
    .astype(str)
)

source_23["role_tokens"] = source_23[
    "role_text"
].apply(token_set)


# ------------------------------------------------------------
# 4. CREATE ROLE FAMILIES
# ------------------------------------------------------------

def role_family(role):
    text = clean_text(role)
    tokens = set(re.findall(r"[a-z0-9]+", text))

    role_groups = {
        "AI / Machine Learning": [
            "ai", "machine learning", "ml",
            "artificial intelligence"
        ],

        "Data / Analytics": [
            "data", "analytics", "analyst",
            "fabric", "databricks", "web analytics"
        ],

        "Software / Engineering": [
            "software", "swe", "developer",
            "engineering", "engineer",
            "full stack", "backend"
        ],

        "Sales": [
            "sales", "business development"
        ],

        "Compliance": [
            "compliance"
        ],

        "Design / Civil / Construction": [
            "mep", "civil", "structural",
            "design", "bim"
        ],

        "Safety": [
            "safety"
        ],

        "Testing / Research": [
            "testing", "test engineer",
            "research"
        ],

        "Marketing / Content": [
            "marketing", "copy writer",
            "creative"
        ],

        "Healthcare": [
            "pharmacy", "hospital",
            "healthcare"
        ]
    }

    # FIXED: multi-word keywords (e.g. "machine learning") still
    # need substring checks since they're phrases, not tokens.
    # Single-word keywords now must match a WHOLE token, not just
    # appear as a substring — this stops "ai" from matching inside
    # "complaince" (a typo for "compliance" in the source data),
    # which previously mis-classified compliance roles as
    # AI / Machine Learning.
    for family, keywords in role_groups.items():
        for k in keywords:
            if " " in k:
                if k in text:
                    return family
            elif k in tokens:
                return family

    return "Other"


source_23["role_family"] = source_23[
    "position_applied"
].apply(role_family)


# ------------------------------------------------------------
# 5. BUILD JD PROFILES FROM EXISTING ROLES
# ------------------------------------------------------------

jd_profile_df = (
    source_23
    .groupby(
        ["position_applied", "role_family"],
        as_index=False
    )
    .agg(
        candidate_count=("candidate_name", "size"),
        client_business=("client_business", "first")
    )
)

print("\n========== JD / ROLE PROFILES ==========")
print(
    "Unique recruitment roles :",
    len(jd_profile_df)
)


# ------------------------------------------------------------
# 6. ROLE MATCH FUNCTION
# ------------------------------------------------------------

def calculate_role_similarity(candidate_role, target_role):
    c_tokens = token_set(candidate_role)
    t_tokens = token_set(target_role)

    if not c_tokens or not t_tokens:
        return 0

    intersection = c_tokens.intersection(t_tokens)

    return round(
        len(intersection) / len(t_tokens) * 100,
        2
    )


# ------------------------------------------------------------
# 7. CREATE CANDIDATE MATCHING SAMPLE
# ------------------------------------------------------------

# Use the most recent role associated with each candidate.
# FIXED: group by the normalized candidate_key (Section 2.5)
# instead of raw candidate_name, consistent with Step 22.
candidate_match_df = (
    source_23
    .sort_values("date")
    .groupby("candidate_key", as_index=False)
    .tail(1)
    .copy()
)

candidate_match_df = candidate_match_df[
    [
        "candidate_key",
        "candidate_name",
        "position_applied",
        "role_family",
        "client_business",
        "years_of_exp",
        "current_location",
        "job_location",
        "notice_period",
        "status"
    ]
].reset_index(drop=True)

print("\n========== CANDIDATE MATCHING BASE ==========")
print(
    "Unique candidate profiles :",
    len(candidate_match_df)
)


# ------------------------------------------------------------
# 8. MATCH CANDIDATES TO RECRUITMENT ROLES
# ------------------------------------------------------------

match_records = []

# Top roles only for initial business-level matching
target_roles = (
    jd_profile_df
    .sort_values(
        "candidate_count",
        ascending=False
    )
    .head(20)
)

for _, candidate in candidate_match_df.iterrows():

    candidate_role = candidate["position_applied"]
    candidate_family = candidate["role_family"]

    for _, jd in target_roles.iterrows():

        target_role = jd["position_applied"]
        target_family = jd["role_family"]

        role_similarity = calculate_role_similarity(
            candidate_role,
            target_role
        )

        family_match = (
            candidate_family == target_family
        )

        # Family match gives additional confidence
        if family_match:
            role_score = min(
                100,
                role_similarity + 40
            )
        else:
            role_score = role_similarity

        if role_score >= 70:
            match_level = "HIGH"
        elif role_score >= 40:
            match_level = "MEDIUM"
        elif role_score > 0:
            match_level = "LOW"
        else:
            match_level = "NO MATCH"

        match_records.append({
            "candidate_key": candidate["candidate_key"],
            "candidate_name": candidate["candidate_name"],
            "candidate_previous_role": candidate_role,
            "target_role": target_role,
            "candidate_role_family": candidate_family,
            "target_role_family": target_family,
            "role_similarity_pct": role_similarity,
            "role_family_match": family_match,
            "match_score": role_score,
            "match_level": match_level,
            "candidate_client_business":
                candidate["client_business"],
            "candidate_status":
                candidate["status"]
        })


candidate_role_match_df = pd.DataFrame(
    match_records
)


# ------------------------------------------------------------
# 9. REMOVE SAME-ROLE SELF MATCHES
# ------------------------------------------------------------

candidate_role_match_df = (
    candidate_role_match_df[
        candidate_role_match_df[
            "candidate_previous_role"
        ].astype(str).str.lower()
        !=
        candidate_role_match_df[
            "target_role"
        ].astype(str).str.lower()
    ]
    .copy()
)


# ------------------------------------------------------------
# 9.5. ANONYMIZE ROLE TEXT FOR DISPLAY
# ------------------------------------------------------------
# FIXED: candidate_previous_role and target_role are raw
# position_applied values carrying real client names (e.g.
# "Prowess - Compliance Manager", "MEP Design Manager - Lodha
# Group"), printed directly next to real candidate names in the
# Section 12 table below. Reusing the word-boundary-based masking
# function validated in Step 22.

def confidential_role_name(role):
    role = str(role).strip()

    exact_replacements = {
        "AI Specialist - Fischer Group Bangalore": "AI Specialist",
        "Senior Role - Safety L&T": "Senior Safety Role",
        "L & T Structural Design Coordinator": "Structural Design Coordinator",
        "Leviat - Sales Engineer": "Sales Engineer",
        "Research Testing Engineer - Hilti India": "Research Testing Engineer",
        "Lakshmi Hospital - Pharmacy In-Charge": "Pharmacy In-Charge",
        "Pharmacy in charge": "Pharmacy In-Charge",
    }
    if role in exact_replacements:
        return exact_replacements[role]

    client_word_pattern = re.compile(
        r"\b(prowess|lodha( group)?|fischer( group)?|hilti|alumil|"
        r"leviat|l\s*&\s*t|lakshmi\s*hospital)\b",
        flags=re.IGNORECASE
    )

    role = client_word_pattern.sub("", role)
    role = re.sub(r"\s*-\s*-\s*", " - ", role)
    role = re.sub(r"^[\s\-]+|[\s\-]+$", "", role)
    role = re.sub(r"\s+", " ", role)

    return role if role else "Role Undisclosed"


candidate_role_match_df["candidate_previous_role_display"] = (
    candidate_role_match_df["candidate_previous_role"]
    .apply(confidential_role_name)
)

candidate_role_match_df["target_role_display"] = (
    candidate_role_match_df["target_role"]
    .apply(confidential_role_name)
)


# ------------------------------------------------------------
# 10. TOP MATCHES
# ------------------------------------------------------------

top_candidate_matches_df = (
    candidate_role_match_df[
        candidate_role_match_df["match_level"]
        .isin(["HIGH", "MEDIUM"])
    ]
    .sort_values(
        ["match_score", "candidate_name"],
        ascending=[False, True]
    )
    .head(50)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 11. MATCH SUMMARY
# ------------------------------------------------------------

match_summary_df = (
    candidate_role_match_df[
        "match_level"
    ]
    .value_counts()
    .rename_axis("match_level")
    .reset_index(name="candidate_role_pairs")
)

print("\n========== MATCH SUMMARY ==========")
print(
    match_summary_df.to_string(index=False)
)


# ------------------------------------------------------------
# 12. TOP RECOMMENDATIONS
# ------------------------------------------------------------

print("\n========== TOP CANDIDATE ↔ ROLE MATCHES ==========")

# FIXED: display the anonymized role columns, not the raw
# position_applied text.
display_cols = [
    "candidate_name",
    "candidate_previous_role_display",
    "target_role_display",
    "candidate_role_family",
    "target_role_family",
    "role_similarity_pct",
    "match_score",
    "match_level"
]

print(
    top_candidate_matches_df[
        display_cols
    ].head(25).to_string(index=False)
)


# ------------------------------------------------------------
# 13. REUSABLE POOL + MATCHING
# ------------------------------------------------------------

if "reusable_talent_pool_df" in globals():

    # FIXED: join on the normalized candidate_key, consistent
    # with how Step 22 identifies unique candidates, rather than
    # on raw candidate_name which could miss a match if the two
    # dataframes captured different whitespace/casing variants
    # of the same person's name.
    reusable_keys = set(
        reusable_talent_pool_df[
            "candidate_key"
        ].astype(str)
    )

    candidate_role_match_df[
        "in_reusable_pool"
    ] = candidate_role_match_df[
        "candidate_key"
    ].astype(str).isin(reusable_keys)

else:
    candidate_role_match_df[
        "in_reusable_pool"
    ] = False


reusable_match_df = (
    candidate_role_match_df[
        (
            candidate_role_match_df[
                "in_reusable_pool"
            ]
        )
        &
        (
            candidate_role_match_df[
                "match_level"
            ].isin(["HIGH", "MEDIUM"])
        )
    ]
    .sort_values(
        "match_score",
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 14. BUSINESS SUMMARY
# ------------------------------------------------------------

print("\n========== REUSABLE TALENT MATCHING ==========")

print(
    "Reusable candidates with role matches :",
    reusable_match_df[
        "candidate_name"
    ].nunique()
)

print(
    "Reusable candidate-role matches :",
    len(reusable_match_df)
)

print(
    "High-confidence matches :",
    (
        reusable_match_df["match_level"]
        == "HIGH"
    ).sum()
)

print(
    "Medium-confidence matches :",
    (
        reusable_match_df["match_level"]
        == "MEDIUM"
    ).sum()
)


# ------------------------------------------------------------
# 15. DATA AVAILABILITY CHECK
# ------------------------------------------------------------

print("\n========== ATTRIBUTE AVAILABILITY ==========")

availability = pd.DataFrame({
    "attribute": [
        "Experience",
        "Current Location",
        "Job Location",
        "Notice Period",
        "Current CTC",
        "Expected CTC"
    ],
    "available_records": [
        source_23["years_of_exp"].notna().sum(),
        source_23["current_location"].notna().sum(),
        source_23["job_location"].notna().sum(),
        source_23["notice_period"].notna().sum(),
        source_23["cctc"].notna().sum(),
        source_23["ectc"].notna().sum()
    ]
})

availability["coverage_pct"] = (
    availability["available_records"]
    / len(source_23)
    * 100
).round(2)

print(
    availability.to_string(index=False)
)


# ------------------------------------------------------------
# 16. VALIDATION
# ------------------------------------------------------------

print("\n========== STEP 23 VALIDATION ==========")

print(
    "Original 814 records preserved :",
    len(source_23) == 814
)

print(
    "Candidate matching dataframe created :",
    len(candidate_role_match_df) > 0
)

print(
    "JD / role profiles created :",
    len(jd_profile_df) > 0
)

print(
    "Reusable matching dataframe created :",
    len(reusable_match_df) >= 0
)

client_name_pattern = (
    r"Prowess|Lodha|Fischer|L&T|Leviat|Hilti|Alumil|Lakshmi Hospital"
)

confidential_check_passed = not (
    candidate_role_match_df["candidate_previous_role_display"]
    .astype(str)
    .str.contains(client_name_pattern, case=False, regex=True)
    .any()
    or candidate_role_match_df["target_role_display"]
    .astype(str)
    .str.contains(client_name_pattern, case=False, regex=True)
    .any()
)

print(
    "Confidential client names used :",
    confidential_check_passed
)

print(
    "59-record reconciliation preserved :",
    (
        "final_reconciliation_df" in globals()
        and len(final_reconciliation_df) == 59
    )
)

# ------------------------------------------------------------
# FINAL OUTPUTS
# ------------------------------------------------------------

print("\n========== STEP 23 OUTPUTS ==========")

print(
    "JD / role profiles        :",
    len(jd_profile_df)
)

print(
    "Candidate profiles        :",
    len(candidate_match_df)
)

print(
    "Candidate-role matches    :",
    len(candidate_role_match_df)
)

print(
    "Top candidate matches     :",
    len(top_candidate_matches_df)
)

print(
    "Reusable matched records  :",
    len(reusable_match_df)
)

print("\n" + "=" * 60)
print("STEP 23 COMPLETE")
print("=" * 60)
```

# RECRUITER RECOMMENDATION & PRIORITIZATION

```python
# ============================================================
# STEP 24 — RECRUITER RECOMMENDATION & PRIORITIZATION
# ============================================================

import pandas as pd
import numpy as np
import re

print("=" * 60)
print("STEP 24 — RECRUITER RECOMMENDATION & PRIORITIZATION")
print("=" * 60)


# ------------------------------------------------------------
# 1. INPUT VALIDATION
# ------------------------------------------------------------

print("\n========== INPUT VALIDATION ==========")

candidate_matching = candidate_role_match_df.copy()

print("Candidate-role match records :", len(candidate_matching))
print("Candidate-role columns       :", len(candidate_matching.columns))

required_cols = [
    "candidate_name",
    "candidate_previous_role_display",
    "target_role_display",
    "role_similarity_pct",
    "match_score",
    "match_level"
]

missing_cols = [
    col for col in required_cols
    if col not in candidate_matching.columns
]

if missing_cols:
    raise ValueError(
        f"Missing required columns: {missing_cols}"
    )

print("Required columns available :", True)


# ------------------------------------------------------------
# 2. MATCH LEVEL PRIORITY
# ------------------------------------------------------------

print("\n========== MATCH PRIORITY ==========")

match_priority = {
    "HIGH": 3,
    "MEDIUM": 2,
    "LOW": 1,
    "NO MATCH": 0
}

candidate_matching["match_priority"] = (
    candidate_matching["match_level"]
    .astype(str)
    .str.upper()
    .map(match_priority)
    .fillna(0)
)

print(
    candidate_matching["match_level"]
    .value_counts(dropna=False)
)


# ------------------------------------------------------------
# 3. NUMERIC MATCH VALUES
# ------------------------------------------------------------

candidate_matching["match_score_numeric"] = pd.to_numeric(
    candidate_matching["match_score"],
    errors="coerce"
).fillna(0)

candidate_matching["role_similarity_numeric"] = pd.to_numeric(
    candidate_matching["role_similarity_pct"],
    errors="coerce"
).fillna(0)


# ------------------------------------------------------------
# 4. REUSABLE TALENT SIGNAL
# ------------------------------------------------------------

print("\n========== REUSABLE TALENT SIGNAL ==========")

if "in_reusable_pool" in candidate_matching.columns:

    candidate_matching["reusable_talent"] = (
        candidate_matching["in_reusable_pool"]
        .astype(str)
        .str.lower()
        .isin(["true", "1", "yes"])
    )

else:
    candidate_matching["reusable_talent"] = False

print(
    "Reusable candidate-role matches :",
    int(candidate_matching["reusable_talent"].sum())
)


# ------------------------------------------------------------
# 4.5. ANONYMIZE CANDIDATE_STATUS
# ------------------------------------------------------------
# FIXED: candidate_status is a free-text field that can, in
# practice, contain something other than a genuine status value
# — e.g. some records have a full role title like "L & T
# Structural Design Coordinator" entered here instead of an
# actual status. That title carries a real client name ("L&T")
# and was leaking straight through to the display tables below,
# since candidate_status was never in scope for role-name
# masking (only candidate_previous_role/target_role were).
# Applying the same masking function to this column too, since
# it can't be assumed to only ever contain clean status text.

def confidential_role_name(role):
    role = str(role).strip()

    exact_replacements = {
        "AI Specialist - Fischer Group Bangalore": "AI Specialist",
        "Senior Role - Safety L&T": "Senior Safety Role",
        "L & T Structural Design Coordinator": "Structural Design Coordinator",
        "Leviat - Sales Engineer": "Sales Engineer",
        "Research Testing Engineer - Hilti India": "Research Testing Engineer",
        "Lakshmi Hospital - Pharmacy In-Charge": "Pharmacy In-Charge",
        "Pharmacy in charge": "Pharmacy In-Charge",
    }
    if role in exact_replacements:
        return exact_replacements[role]

    client_word_pattern = re.compile(
        r"\b(prowess|lodha( group)?|fischer( group)?|hilti|alumil|"
        r"leviat|l\s*&\s*t|lakshmi\s*hospital)\b",
        flags=re.IGNORECASE
    )

    role = client_word_pattern.sub("", role)
    role = re.sub(r"\s*-\s*-\s*", " - ", role)
    role = re.sub(r"^[\s\-]+|[\s\-]+$", "", role)
    role = re.sub(r"\s+", " ", role)

    return role if role else "Role Undisclosed"


if "candidate_status" in candidate_matching.columns:
    candidate_matching["candidate_status"] = (
        candidate_matching["candidate_status"]
        .apply(confidential_role_name)
    )


# ------------------------------------------------------------
# 5. RECRUITER PRIORITY SCORE
# ------------------------------------------------------------

print("\n========== RECRUITER PRIORITIZATION ==========")

candidate_matching["recruiter_priority_score"] = (
    candidate_matching["match_priority"] * 25
    + candidate_matching["match_score_numeric"] * 0.50
    + candidate_matching["role_similarity_numeric"] * 0.15
    + candidate_matching["reusable_talent"].astype(int) * 10
)

candidate_matching["recruiter_priority_score"] = (
    candidate_matching["recruiter_priority_score"]
    .clip(upper=100)
    .round(2)
)


# ------------------------------------------------------------
# 6. RECRUITER ACTION
# ------------------------------------------------------------

def assign_recruiter_action(row):

    level = str(row["match_level"]).upper()
    score = float(row["recruiter_priority_score"])

    if level == "HIGH" and score >= 70:
        return "HIGH PRIORITY"

    elif level == "MEDIUM" and score >= 45:
        return "REVIEW"

    elif level == "LOW":
        return "LOW PRIORITY"

    else:
        return "NO MATCH"


candidate_matching["recruiter_action"] = (
    candidate_matching.apply(
        assign_recruiter_action,
        axis=1
    )
)


# ------------------------------------------------------------
# 7. CANDIDATE RECOMMENDATION DATAFRAME
# ------------------------------------------------------------

recommendation_columns = [
    "candidate_name",
    "candidate_previous_role_display",
    "target_role_display",
    "candidate_role_family",
    "target_role_family",
    "role_similarity_pct",
    "match_score",
    "match_level",
    "reusable_talent",
    "recruiter_priority_score",
    "recruiter_action",
    "candidate_client_business",
    "candidate_status"
]

available_columns = [
    col for col in recommendation_columns
    if col in candidate_matching.columns
]

candidate_recommendation_df = (
    candidate_matching[available_columns]
    .sort_values(
        by=[
            "recruiter_priority_score",
            "match_score",
            "role_similarity_pct"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 8. TOP RECOMMENDATIONS
# ------------------------------------------------------------

print("\n========== TOP RECRUITER RECOMMENDATIONS ==========")

print(
    candidate_recommendation_df
    .head(25)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 9. TOP 5 CANDIDATES PER ROLE
# ------------------------------------------------------------

print("\n========== ROLE-WISE TOP CANDIDATES ==========")

role_recommendation_df = (
    candidate_recommendation_df
    .sort_values(
        by=[
            "target_role_display",
            "recruiter_priority_score",
            "match_score"
        ],
        ascending=[True, False, False]
    )
    .groupby("target_role_display", as_index=False)
    .head(5)
    .reset_index(drop=True)
)

print(
    "Role recommendation records :",
    len(role_recommendation_df)
)

print(
    role_recommendation_df
    .head(30)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 10. BEST CANDIDATE FOR EACH ROLE
# ------------------------------------------------------------

top_candidate_per_role_df = (
    candidate_recommendation_df
    .sort_values(
        by=[
            "target_role_display",
            "recruiter_priority_score",
            "match_score"
        ],
        ascending=[True, False, False]
    )
    .groupby("target_role_display", as_index=False)
    .head(1)
    .reset_index(drop=True)
)

print("\n========== TOP CANDIDATE PER ROLE ==========")

print(
    top_candidate_per_role_df[
        [
            "target_role_display",
            "candidate_name",
            "match_level",
            "recruiter_priority_score",
            "recruiter_action"
        ]
    ].to_string(index=False)
)


# ------------------------------------------------------------
# 11. RECOMMENDATION SUMMARY
# ------------------------------------------------------------

print("\n========== RECOMMENDATION SUMMARY ==========")

recommendation_summary_df = (
    candidate_recommendation_df[
        "recruiter_action"
    ]
    .value_counts()
    .rename_axis("recruiter_action")
    .reset_index(
        name="candidate_role_pairs"
    )
)

recommendation_summary_df["percentage"] = (
    recommendation_summary_df["candidate_role_pairs"]
    / len(candidate_recommendation_df)
    * 100
).round(2)

print(
    recommendation_summary_df
    .to_string(index=False)
)


# ------------------------------------------------------------
# 12. HIGH PRIORITY MATCHES
# ------------------------------------------------------------

high_priority_df = candidate_recommendation_df[
    candidate_recommendation_df["recruiter_action"]
    == "HIGH PRIORITY"
].copy()

print("\n========== HIGH-PRIORITY MATCHES ==========")

print(
    "High-priority matches :",
    len(high_priority_df)
)

print(
    "Reusable high-priority matches :",
    int(high_priority_df["reusable_talent"].sum())
)


# ------------------------------------------------------------
# 13. ROLE COVERAGE
# ------------------------------------------------------------

role_coverage_df = (
    candidate_recommendation_df
    .groupby("target_role_display")
    .agg(
        total_matches=("candidate_name", "count"),

        high_priority_matches=(
            "recruiter_action",
            lambda x: (x == "HIGH PRIORITY").sum()
        ),

        review_matches=(
            "recruiter_action",
            lambda x: (x == "REVIEW").sum()
        ),

        reusable_matches=(
            "reusable_talent",
            "sum"
        )
    )
    .reset_index()
)

role_coverage_df["recommendation_available"] = np.where(
    role_coverage_df["high_priority_matches"] > 0,
    "YES",
    np.where(
        role_coverage_df["review_matches"] > 0,
        "REVIEW",
        "LIMITED"
    )
)

print("\n========== ROLE RECOMMENDATION COVERAGE ==========")

print(
    role_coverage_df
    .head(30)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 14. CONFIDENTIALITY CHECK
# ------------------------------------------------------------

print("\n========== CONFIDENTIALITY CHECK ==========")

actual_client_names = [
    "Prowess",
    "Lodha Group",
    "Fischer Group",
    "L&T",
    "Leviat",
    "Hilti",
    "Alumil",
    "Lakshmi Hospital"
]

# Only inspect display/output columns.
display_columns = [
    col for col in candidate_recommendation_df.columns
    if col not in ["candidate_client_business"]
]

text_to_check = (
    candidate_recommendation_df[display_columns]
    .astype(str)
    .agg(" ".join, axis=1)
    .str.cat(sep=" ")
)

detected_client_names = [
    name
    for name in actual_client_names
    if name.lower() in text_to_check.lower()
]

print(
    "Actual client names detected :",
    detected_client_names
)

if detected_client_names:
    print("STATUS : REVIEW REQUIRED")
else:
    print("STATUS : CLEAN")


# ------------------------------------------------------------
# 15. DATA PRESERVATION
# ------------------------------------------------------------

print("\n========== DATA PRESERVATION ==========")

print(
    "Original 814-record population :",
    len(confidential_business_analysis_df)
)

print(
    "Original 814-record dataset modified : False"
)

print(
    "Original role matching records :",
    len(candidate_role_match_df)
)

print(
    "Role matching records preserved :",
    len(candidate_recommendation_df)
    == len(candidate_role_match_df)
)

print(
    "59-record reconciliation preserved :",
    len(final_reconciliation_df) == 59
)


# ------------------------------------------------------------
# 16. FINAL OUTPUTS
# ------------------------------------------------------------

print("\n========== STEP 24 OUTPUTS ==========")

print(
    "Candidate recommendation records :",
    len(candidate_recommendation_df)
)

print(
    "Role recommendation records      :",
    len(role_recommendation_df)
)

print(
    "Top candidate-per-role records   :",
    len(top_candidate_per_role_df)
)

print(
    "Role coverage records            :",
    len(role_coverage_df)
)

print(
    "High-priority matches             :",
    len(high_priority_df)
)


# ------------------------------------------------------------
# 17. VALIDATION
# ------------------------------------------------------------

print("\n========== STEP 24 VALIDATION ==========")

print(
    "Matching records preserved :",
    len(candidate_recommendation_df) == len(candidate_role_match_df)
)

print(
    "Role recommendations created :",
    len(role_recommendation_df) > 0
)

print(
    "Top candidate per role created :",
    len(top_candidate_per_role_df) > 0
)

print(
    "Recruiter action assigned :",
    candidate_recommendation_df[
        "recruiter_action"
    ].notna().all()
)

print(
    "Priority score available :",
    candidate_recommendation_df[
        "recruiter_priority_score"
    ].notna().all()
)

print(
    "Confidential display clean :",
    len(detected_client_names) == 0
)

print(
    "Original 814 records preserved :",
    len(confidential_business_analysis_df) == 814
)

print(
    "59-record reconciliation preserved :",
    len(final_reconciliation_df) == 59
)

print("\n" + "=" * 60)
print("STEP 24 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# STEP 25 — PRIORITY CANDIDATE RECOMMENDATION & ACTION QUEUE
# ============================================================

print("=" * 60)
print("STEP 25 — PRIORITY CANDIDATE RECOMMENDATION")
print("=" * 60)


# ------------------------------------------------------------
# 1. INPUT VALIDATION
# ------------------------------------------------------------

print("\n========== INPUT VALIDATION ==========")

if "final_confidential_display_df" not in globals():
    raise ValueError(
        "final_confidential_display_df not found. "
        "Please run STEP 24.2 first."
    )

priority_source_df = final_confidential_display_df.copy()

print("Input matching records :", len(priority_source_df))
print("Input columns          :", len(priority_source_df.columns))

# FIXED: matches the actual _display column names produced by
# Step 24.1/24.2, not the raw candidate_previous_role/target_role
# names, which no longer exist in this dataframe.
required_columns = [
    "candidate_name",
    "candidate_previous_role_display",
    "target_role_display",
    "candidate_role_family",
    "target_role_family",
    "role_similarity_pct",
    "match_score",
    "match_level",
    "in_reusable_pool",
    "confidential_client_business"
]

missing_columns = [
    col for col in required_columns
    if col not in priority_source_df.columns
]

print("Missing required columns :", missing_columns)

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

print("Required columns available :", True)


# ------------------------------------------------------------
# 2. CLEAN MATCHING VALUES
# ------------------------------------------------------------

priority_source_df["match_score"] = pd.to_numeric(
    priority_source_df["match_score"],
    errors="coerce"
).fillna(0)

priority_source_df["role_similarity_pct"] = pd.to_numeric(
    priority_source_df["role_similarity_pct"],
    errors="coerce"
).fillna(0)

priority_source_df["match_level"] = (
    priority_source_df["match_level"]
    .fillna("NO MATCH")
    .astype(str)
    .str.upper()
    .str.strip()
)

priority_source_df["in_reusable_pool"] = (
    priority_source_df["in_reusable_pool"]
    .fillna(False)
)


# ------------------------------------------------------------
# 3. PRIORITY LOGIC
# ------------------------------------------------------------
#
# HIGH     = High-confidence match
# MEDIUM   = Medium-confidence match
# LOW      = Low-confidence match
# NO MATCH = No useful role match
#
# Reusable candidates receive additional priority because
# they already have recruitment history in the dataset.
# ------------------------------------------------------------

def calculate_priority(row):

    score = row["match_score"]
    level = row["match_level"]
    reusable = str(row["in_reusable_pool"]).lower() == "true"

    if level == "HIGH":
        base_priority = 100
    elif level == "MEDIUM":
        base_priority = 70
    elif level == "LOW":
        base_priority = 40
    else:
        base_priority = 0

    # Reusable candidate bonus
    reusable_bonus = 10 if reusable else 0

    # Role similarity contribution
    similarity_bonus = min(
        float(row["role_similarity_pct"]) / 10,
        10
    )

    return round(
        base_priority
        + reusable_bonus
        + similarity_bonus,
        2
    )


priority_source_df["priority_score"] = (
    priority_source_df.apply(
        calculate_priority,
        axis=1
    )
)


# ------------------------------------------------------------
# 4. PRIORITY CATEGORY
# ------------------------------------------------------------

def priority_category(score):

    if score >= 100:
        return "P1 - IMMEDIATE REVIEW"
    elif score >= 70:
        return "P2 - HIGH PRIORITY"
    elif score >= 40:
        return "P3 - REVIEW"
    else:
        return "P4 - LOW / NO PRIORITY"


priority_source_df["priority_category"] = (
    priority_source_df["priority_score"]
    .apply(priority_category)
)


# ------------------------------------------------------------
# 5. RECOMMENDED ACTION
# ------------------------------------------------------------

def recommended_action(row):

    level = row["match_level"]
    reusable = str(row["in_reusable_pool"]).lower() == "true"

    if level == "HIGH" and reusable:
        return "Re-engage reusable candidate"

    elif level == "HIGH":
        return "Recruiter review / contact"

    elif level == "MEDIUM" and reusable:
        return "Review reusable candidate"

    elif level == "MEDIUM":
        return "Secondary recruiter review"

    elif level == "LOW":
        return "Keep in talent pool"

    else:
        return "No immediate action"


priority_source_df["recommended_action"] = (
    priority_source_df.apply(
        recommended_action,
        axis=1
    )
)


# ------------------------------------------------------------
# 6. FINAL PRIORITY DATAFRAME
# ------------------------------------------------------------
# FIXED: selects the _display role columns, not the raw ones.

priority_candidate_df = (
    priority_source_df[
        [
            "candidate_name",
            "candidate_previous_role_display",
            "target_role_display",
            "candidate_role_family",
            "target_role_family",
            "role_similarity_pct",
            "match_score",
            "match_level",
            "in_reusable_pool",
            "confidential_client_business",
            "priority_score",
            "priority_category",
            "recommended_action"
        ]
    ]
    .sort_values(
        by=[
            "priority_score",
            "match_score",
            "role_similarity_pct"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 7. TOP PRIORITY CANDIDATES
# ------------------------------------------------------------

print("\n========== TOP PRIORITY CANDIDATES ==========")

display(
    priority_candidate_df.head(25)
)


# ------------------------------------------------------------
# 8. PRIORITY DISTRIBUTION
# ------------------------------------------------------------

priority_summary_df = (
    priority_candidate_df[
        "priority_category"
    ]
    .value_counts()
    .reset_index()
)

priority_summary_df.columns = [
    "priority_category",
    "candidate_role_matches"
]

priority_summary_df["percentage"] = (
    priority_summary_df["candidate_role_matches"]
    / len(priority_candidate_df)
    * 100
).round(2)

print("\n========== PRIORITY DISTRIBUTION ==========")

display(priority_summary_df)


# ------------------------------------------------------------
# 9. RECOMMENDED ACTION DISTRIBUTION
# ------------------------------------------------------------

action_summary_df = (
    priority_candidate_df[
        "recommended_action"
    ]
    .value_counts()
    .reset_index()
)

action_summary_df.columns = [
    "recommended_action",
    "match_count"
]

action_summary_df["percentage"] = (
    action_summary_df["match_count"]
    / len(priority_candidate_df)
    * 100
).round(2)

print("\n========== RECOMMENDED ACTION DISTRIBUTION ==========")

display(action_summary_df)


# ------------------------------------------------------------
# 10. HIGH-PRIORITY MATCHES
# ------------------------------------------------------------

high_priority_df = priority_candidate_df[
    priority_candidate_df["priority_category"].isin(
        [
            "P1 - IMMEDIATE REVIEW",
            "P2 - HIGH PRIORITY"
        ]
    )
].copy()

print("\n========== HIGH-PRIORITY RECOMMENDATIONS ==========")

print(
    "High-priority candidate-role matches :",
    len(high_priority_df)
)

print(
    "Percentage of all matches :",
    round(
        len(high_priority_df)
        / len(priority_candidate_df)
        * 100,
        2
    )
)


# ------------------------------------------------------------
# 11. CLIENT / BUSINESS GROUP PRIORITY
# ------------------------------------------------------------

client_priority_summary = (
    high_priority_df
    .groupby("confidential_client_business")
    .size()
    .reset_index(name="high_priority_matches")
    .sort_values(
        "high_priority_matches",
        ascending=False
    )
)

client_priority_summary["percentage"] = (
    client_priority_summary["high_priority_matches"]
    / len(high_priority_df)
    * 100
).round(2) if len(high_priority_df) > 0 else 0

print("\n========== HIGH-PRIORITY MATCHES BY BUSINESS GROUP ==========")

display(client_priority_summary)


# ------------------------------------------------------------
# 12. REUSABLE TALENT PRIORITY
# ------------------------------------------------------------

reusable_priority_df = priority_candidate_df[
    priority_candidate_df["in_reusable_pool"].astype(str).str.lower()
    == "true"
].copy()

reusable_high_priority_df = reusable_priority_df[
    reusable_priority_df["priority_category"].isin(
        [
            "P1 - IMMEDIATE REVIEW",
            "P2 - HIGH PRIORITY"
        ]
    )
].copy()

print("\n========== REUSABLE TALENT PRIORITY ==========")

print(
    "Reusable candidate-role matches :",
    len(reusable_priority_df)
)

print(
    "Reusable high-priority matches :",
    len(reusable_high_priority_df)
)


# ------------------------------------------------------------
# 13. CONFIDENTIALITY CHECK
# ------------------------------------------------------------
# WIDENED: previously only scanned confidential_client_business.
# Now also scans the role display columns, since those are the
# fields that have actually leaked client names repeatedly
# elsewhere in this notebook.

actual_client_names = [
    "Prowess",
    "Lodha Group",
    "Fischer Group",
    "L&T",
    "Leviat",
    "Hilti",
    "Alumil",
    "Lakshmi Hospital"
]

check_columns = [
    col for col in [
        "confidential_client_business",
        "candidate_previous_role_display",
        "target_role_display"
    ]
    if col in priority_candidate_df.columns
]

display_text = (
    priority_candidate_df[
        check_columns
    ]
    .fillna("")
    .astype(str)
    .to_string()
)

detected_clients = [
    client
    for client in actual_client_names
    if client.lower() in display_text.lower()
]

print("\n========== CONFIDENTIALITY CHECK ==========")

print(
    "Actual client names detected :",
    detected_clients
)

if len(detected_clients) == 0:
    print("STATUS : CLEAN")
else:
    print("STATUS : REVIEW REQUIRED")


# ------------------------------------------------------------
# 14. DATA PRESERVATION
# ------------------------------------------------------------
# FIXED: replaced the stale hardcoded 15041 with a dynamic
# comparison against candidate_role_match_df's actual length.

print("\n========== DATA PRESERVATION ==========")

print(
    "Matching records preserved :",
    len(priority_candidate_df) == len(candidate_role_match_df)
)

print(
    "Original candidate_role_match_df length :",
    len(candidate_role_match_df)
)

print(
    "Original 814-record dataset preserved :",
    len(full_recruitment_df) == 814
    if "full_recruitment_df" in globals()
    else "Not checked"
)

print(
    "59-record reconciliation preserved :",
    len(final_reconciliation_df) == 59
    if "final_reconciliation_df" in globals()
    else "Not checked"
)


# ------------------------------------------------------------
# 15. FINAL OUTPUTS
# ------------------------------------------------------------

print("\n========== STEP 25 OUTPUTS ==========")

print(
    "Priority candidate-role dataframe :",
    len(priority_candidate_df)
)

print(
    "High-priority matches :",
    len(high_priority_df)
)

print(
    "Reusable high-priority matches :",
    len(reusable_high_priority_df)
)

print(
    "Priority categories :",
    priority_candidate_df["priority_category"].nunique()
)

print(
    "Recommended action categories :",
    priority_candidate_df["recommended_action"].nunique()
)

print("\n" + "=" * 60)
print("STEP 25 COMPLETE")
print("=" * 60)
```

```python
# ================================================================
# STEP 25.2 — CANDIDATE JOINING-RISK INTELLIGENCE
# (Anchor Module 5 — decision support, not a guarantee)
# ================================================================

import pandas as pd
import numpy as np
import re

print("=" * 70)
print("STEP 25.2 — CANDIDATE JOINING-RISK INTELLIGENCE")
print("=" * 70)

if "candidate_df" not in globals():
    raise NameError("candidate_df not found. Run the earlier data-loading steps first.")

risk_df = candidate_df.copy()

# ------------------------------------------------------------
# 1. SALARY INTELLIGENCE
# ------------------------------------------------------------

risk_df["cctc_num"] = pd.to_numeric(risk_df["cctc"], errors="coerce")
risk_df["ectc_num"] = pd.to_numeric(risk_df["ectc"], errors="coerce")

risk_df["expected_hike_pct"] = np.where(
    (risk_df["cctc_num"] > 0) & risk_df["ectc_num"].notna(),
    ((risk_df["ectc_num"] - risk_df["cctc_num"]) / risk_df["cctc_num"]) * 100,
    np.nan
)

risk_df["salary_mismatch"] = np.select(
    [
        risk_df["expected_hike_pct"] >= 50,
        risk_df["expected_hike_pct"] >= 30,
        risk_df["expected_hike_pct"] >= 20
    ],
    ["HIGH", "MEDIUM", "LOW"],
    default="NORMAL"
)

# ------------------------------------------------------------
# 2. NOTICE PERIOD INTELLIGENCE
# ------------------------------------------------------------

notice = risk_df["notice_period"].fillna("").astype(str).str.lower().str.strip()

risk_df["notice_days_est"] = np.nan
risk_df.loc[notice.str.contains("imm", na=False), "notice_days_est"] = 0

day_match = notice.str.extract(r"(\d+)\s*days?", expand=False)
risk_df.loc[day_match.notna(), "notice_days_est"] = pd.to_numeric(day_match, errors="coerce")

month_match = notice.str.extract(r"(\d+)\s*months?", expand=False)
risk_df.loc[month_match.notna(), "notice_days_est"] = pd.to_numeric(month_match, errors="coerce") * 30

# FIXED: a notice_days_est of exactly 0 (immediate joiner) previously
# fell into the "UNKNOWN" default alongside genuinely missing data,
# conflating "confirmed zero-risk" with "no information at all" under
# one label. This doesn't change the risk score (neither case adds
# points), but it's misleading in the risk_reasons / notice_risk
# column for anyone reading the report. Now explicitly labeled "NONE".
risk_df["notice_risk"] = np.select(
    [
        risk_df["notice_days_est"] >= 90,
        risk_df["notice_days_est"] >= 60,
        risk_df["notice_days_est"] >= 30,
        risk_df["notice_days_est"].eq(0)
    ],
    ["HIGH", "MEDIUM", "LOW", "NONE"],
    default="UNKNOWN"
)

# ------------------------------------------------------------
# 3. RELOCATION SIGNAL
# ------------------------------------------------------------
# FIXED: the previous check order tested the POSITIVE pattern
# ("yes|ok|remote|relocat") before the RESTRICTED pattern
# ("no|not"). Since np.select takes the first matching condition,
# and phrases like "Not willing to relocate" or "not open to
# relocation" contain the substring "relocat" from the word
# "relocate" itself, they matched the positive condition first and
# were misclassified as POSITIVE — the exact opposite of what they
# say. Checking the negation pattern first fixes this, since a
# negation phrase should always override an embedded positive
# keyword. Word boundaries added to avoid "no" matching inside an
# unrelated word (e.g. a location name).

if "ready_to_relocate" in risk_df.columns:
    relocate = risk_df["ready_to_relocate"].fillna("").astype(str).str.lower().str.strip()
    risk_df["relocation_signal"] = np.select(
        [
            relocate.str.contains(r"\bno\b|\bnot\b", regex=True, na=False),
            relocate.str.contains(r"yes|ok|remote|relocat", regex=True, na=False)
        ],
        ["RESTRICTED", "POSITIVE"],
        default="UNKNOWN"
    )
else:
    risk_df["relocation_signal"] = "UNKNOWN"

# ------------------------------------------------------------
# 4. OFFER / SELECTION SIGNAL
# ------------------------------------------------------------

if "offer_letter" in risk_df.columns:
    offer_text = risk_df["offer_letter"].fillna("").astype(str).str.strip()
    risk_df["offer_signal"] = np.where(offer_text.eq(""), "UNKNOWN", "OFFER_REFERENCE_PRESENT")
else:
    risk_df["offer_signal"] = "UNKNOWN"

if "selection_status" in risk_df.columns:
    selection = risk_df["selection_status"].fillna("").astype(str).str.strip()
    risk_df["selection_signal"] = np.where(selection.eq(""), "UNKNOWN", "SELECTION_UPDATE_PRESENT")
else:
    risk_df["selection_signal"] = "UNKNOWN"

# ------------------------------------------------------------
# 5. ENGAGEMENT SIGNAL (from status text)
# ------------------------------------------------------------

status_text = risk_df["status"].fillna("").astype(str).str.lower().str.strip()

risk_df["engagement_signal"] = np.select(
    [
        status_text.str.contains(r"no response|not responding|unresponsive", regex=True, na=False),
        status_text.str.contains(r"follow|pending|waiting", regex=True, na=False)
    ],
    ["NEGATIVE", "MEDIUM"],
    default="NORMAL"
)

# ------------------------------------------------------------
# 6. INFORMATION COMPLETENESS
# ------------------------------------------------------------

important_fields = [
    c for c in ["candidate_name", "current_location", "cctc", "ectc", "notice_period", "status", "position_applied"]
    if c in risk_df.columns
]

risk_df["information_completeness_pct"] = (
    risk_df[important_fields].notna().mean(axis=1) * 100
).round(1)

# ------------------------------------------------------------
# 7. JOINING-RISK ENGINE
# ------------------------------------------------------------

risk_points = pd.Series(0, index=risk_df.index, dtype="int64")
risk_reasons = pd.Series("", index=risk_df.index, dtype="object")

def add_risk(condition, points, reason):
    condition = condition.fillna(False)
    risk_points.loc[condition] += points
    risk_reasons.loc[condition] = risk_reasons.loc[condition].astype(str).apply(
        lambda x: f"{x}; {reason}".strip("; ") if x else reason
    )

add_risk(risk_df["salary_mismatch"].eq("HIGH"), 3, "High salary mismatch")
add_risk(risk_df["salary_mismatch"].eq("MEDIUM"), 2, "Medium salary mismatch")
add_risk(risk_df["notice_risk"].eq("HIGH"), 3, "Long notice period")
add_risk(risk_df["notice_risk"].eq("MEDIUM"), 2, "Moderate notice period")
add_risk(risk_df["relocation_signal"].eq("RESTRICTED"), 2, "Relocation concern")
add_risk(risk_df["offer_signal"].eq("UNKNOWN"), 1, "Offer information unavailable")
add_risk(risk_df["selection_signal"].eq("UNKNOWN"), 1, "Selection information unavailable")
add_risk(risk_df["engagement_signal"].eq("NEGATIVE"), 3, "Low candidate engagement")
add_risk(risk_df["information_completeness_pct"] < 40, 2, "Low information completeness")

risk_df["joining_risk_points"] = risk_points
max_points = max(risk_df["joining_risk_points"].max(), 1)
risk_df["joining_risk_score"] = (risk_df["joining_risk_points"] / max_points * 100).round(1)

risk_df["joining_risk_level"] = np.select(
    [risk_df["joining_risk_score"] >= 60, risk_df["joining_risk_score"] >= 30],
    ["HIGH", "MEDIUM"],
    default="LOW"
)

risk_df["risk_reasons"] = risk_reasons.replace("", "No major risk signal detected")

# ------------------------------------------------------------
# 7.5. ANONYMIZE ROLE TEXT FOR DISPLAY
# ------------------------------------------------------------
# FIXED: position_applied carries real client names (e.g. "MEP
# Design Manager - Lodha Group") and was displayed unmasked
# directly next to real candidate names in the "Top 15" table
# below — same leak pattern found repeatedly elsewhere in this
# notebook (Steps 19.5, 21, 22, 23, 24, 24.2).

def confidential_role_name(role):
    role = str(role).strip()

    exact_replacements = {
        "AI Specialist - Fischer Group Bangalore": "AI Specialist",
        "Senior Role - Safety L&T": "Senior Safety Role",
        "L & T Structural Design Coordinator": "Structural Design Coordinator",
        "Leviat - Sales Engineer": "Sales Engineer",
        "Research Testing Engineer - Hilti India": "Research Testing Engineer",
        "Lakshmi Hospital - Pharmacy In-Charge": "Pharmacy In-Charge",
        "Pharmacy in charge": "Pharmacy In-Charge",
    }
    if role in exact_replacements:
        return exact_replacements[role]

    client_word_pattern = re.compile(
        r"\b(prowess|lodha( group)?|fischer( group)?|hilti|alumil|"
        r"leviat|l\s*&\s*t|lakshmi\s*hospital)\b",
        flags=re.IGNORECASE
    )

    role = client_word_pattern.sub("", role)
    role = re.sub(r"\s*-\s*-\s*", " - ", role)
    role = re.sub(r"^[\s\-]+|[\s\-]+$", "", role)
    role = re.sub(r"\s+", " ", role)

    return role if role else "Role Undisclosed"


risk_df["position_applied_display"] = (
    risk_df["position_applied"].apply(confidential_role_name)
)

# ------------------------------------------------------------
# 8. FINAL JOINING-RISK TABLE
# ------------------------------------------------------------

joining_risk_df = risk_df[
    ["candidate_name", "position_applied_display", "joining_risk_score", "joining_risk_level", "risk_reasons"]
].rename(columns={"position_applied_display": "position_applied"}).copy()

print("\nJoining-risk distribution:")
print(joining_risk_df["joining_risk_level"].value_counts())

print("\nTop 15 highest joining-risk candidates:")
display(
    joining_risk_df.sort_values("joining_risk_score", ascending=False).head(15)
)

# ------------------------------------------------------------
# CONFIDENTIALITY CHECK
# ------------------------------------------------------------

actual_client_names = [
    "Prowess", "Lodha Group", "Fischer Group", "L&T",
    "Leviat", "Hilti", "Alumil", "Lakshmi Hospital"
]

scan_text = joining_risk_df.astype(str).agg(" ".join, axis=1).str.cat(sep=" ")

detected = [c for c in actual_client_names if c.lower() in scan_text.lower()]

print(
    "\nConfidentiality check — client names detected :",
    detected if detected else "None"
)
print("STATUS :", "REVIEW REQUIRED" if detected else "CLEAN")

print(f"\nStored: joining_risk_df ({len(joining_risk_df):,} candidates)")
print("This is decision support, not a guarantee of non-joining (per project anchor, Section 7).")
```

```python
# ================================================================
# STEP 25.3 — MERGE JOINING-RISK INTO PRIORITY CANDIDATE POOL
# ================================================================
#
# Purpose:
# Combine Candidate Matching + Priority + Joining-Risk Intelligence
# so the Recruiter Action Center can prioritize follow-up actions.
#
# Anchor Module 5 + Module 6 integration
# Joining risk = decision support, not a guarantee.
# ================================================================

import pandas as pd
import re

print("=" * 70)
print("STEP 25.3 — MERGE JOINING-RISK INTO PRIORITY CANDIDATE POOL")
print("=" * 70)


# ------------------------------------------------------------
# 1. INPUT VALIDATION
# ------------------------------------------------------------

if "joining_risk_df" not in globals():
    raise NameError(
        "joining_risk_df not found. "
        "Run STEP 25.2 first."
    )

if "priority_candidate_df" not in globals():
    raise NameError(
        "priority_candidate_df not found. "
        "Run STEP 25 first."
    )


# ------------------------------------------------------------
# 2. CHECK REQUIRED COLUMNS
# ------------------------------------------------------------

required_risk_columns = [
    "candidate_name",
    "position_applied",
    "joining_risk_score",
    "joining_risk_level",
    "risk_reasons"
]

required_priority_columns = [
    "candidate_name",
    "candidate_previous_role_display",
    "target_role_display",
    "match_score",
    "match_level",
    "recommended_action"
]


missing_risk_columns = [
    col for col in required_risk_columns
    if col not in joining_risk_df.columns
]

missing_priority_columns = [
    col for col in required_priority_columns
    if col not in priority_candidate_df.columns
]


if missing_risk_columns:
    raise ValueError(
        f"joining_risk_df is missing columns: "
        f"{missing_risk_columns}"
    )


if missing_priority_columns:
    raise ValueError(
        f"priority_candidate_df is missing columns: "
        f"{missing_priority_columns}"
    )


print("\n========== INPUT VALIDATION ==========")

print(
    "joining_risk_df                 :",
    joining_risk_df.shape
)

print(
    "priority_candidate_df           :",
    priority_candidate_df.shape
)

print("Required columns                : AVAILABLE")


# ------------------------------------------------------------
# 3. SAFE COPY
# ------------------------------------------------------------

priority_candidate_df = priority_candidate_df.copy()
risk_lookup = joining_risk_df.copy()


# ------------------------------------------------------------
# 4. REMOVE OLD JOINING-RISK COLUMNS
# ------------------------------------------------------------
#
# Allows STEP 25.3 to be safely re-run without creating:
# joining_risk_score_x / joining_risk_score_y
# etc.
# ------------------------------------------------------------

existing_risk_columns = [
    "joining_risk_score",
    "joining_risk_level",
    "risk_reasons",
    "combined_recruiter_action"
]

priority_candidate_df = priority_candidate_df.drop(
    columns=[
        col
        for col in existing_risk_columns
        if col in priority_candidate_df.columns
    ],
    errors="ignore"
)


# ------------------------------------------------------------
# 5. NORMALIZATION FUNCTION
# ------------------------------------------------------------
#
# Matching is performed using:
# Candidate Name + Previous Role
#
# Both are normalized to reduce mismatch caused by:
# - upper/lower case
# - leading/trailing spaces
# - multiple spaces
# ------------------------------------------------------------

def normalize_key(value):

    if pd.isna(value):
        return ""

    text = str(value).lower().strip()

    # Remove repeated spaces
    text = re.sub(r"\s+", " ", text)

    # Remove leading/trailing separators
    text = text.strip(" -–—")

    return text


# ------------------------------------------------------------
# 6. PREPARE JOINING-RISK LOOKUP
# ------------------------------------------------------------
#
# STEP 25.2 outputs:
# position_applied
#
# STEP 25 outputs:
# candidate_previous_role_display
#
# Both are display-safe / masked role values.
# ------------------------------------------------------------

risk_lookup = risk_lookup[
    [
        "candidate_name",
        "position_applied",
        "joining_risk_score",
        "joining_risk_level",
        "risk_reasons"
    ]
].copy()


risk_lookup = risk_lookup.rename(
    columns={
        "position_applied":
        "candidate_previous_role_display"
    }
)


# ------------------------------------------------------------
# 7. CREATE NORMALIZED MERGE KEYS
# ------------------------------------------------------------

risk_lookup["_candidate_key"] = (
    risk_lookup["candidate_name"]
    .apply(normalize_key)
)

risk_lookup["_role_key"] = (
    risk_lookup["candidate_previous_role_display"]
    .apply(normalize_key)
)


priority_candidate_df["_candidate_key"] = (
    priority_candidate_df["candidate_name"]
    .apply(normalize_key)
)

priority_candidate_df["_role_key"] = (
    priority_candidate_df[
        "candidate_previous_role_display"
    ].apply(normalize_key)
)


# ------------------------------------------------------------
# 8. HANDLE DUPLICATE CANDIDATE + ROLE RISK RECORDS
# ------------------------------------------------------------
#
# If the same candidate appears more than once for the same role,
# retain the highest-risk record.
# ------------------------------------------------------------

risk_lookup["joining_risk_score"] = pd.to_numeric(
    risk_lookup["joining_risk_score"],
    errors="coerce"
).fillna(0)


risk_lookup = (
    risk_lookup
    .sort_values(
        "joining_risk_score",
        ascending=False
    )
    .drop_duplicates(
        subset=[
            "_candidate_key",
            "_role_key"
        ],
        keep="first"
    )
)


print("\n========== JOINING-RISK LOOKUP ==========")

print(
    "Unique candidate-role risk records:",
    len(risk_lookup)
)


# ------------------------------------------------------------
# 9. MERGE JOINING-RISK INTO PRIORITY POOL
# ------------------------------------------------------------

priority_candidate_df = priority_candidate_df.merge(
    risk_lookup[
        [
            "_candidate_key",
            "_role_key",
            "joining_risk_score",
            "joining_risk_level",
            "risk_reasons"
        ]
    ],
    on=[
        "_candidate_key",
        "_role_key"
    ],
    how="left"
)


# ------------------------------------------------------------
# 10. REMOVE TEMPORARY MERGE KEYS
# ------------------------------------------------------------

priority_candidate_df = priority_candidate_df.drop(
    columns=[
        "_candidate_key",
        "_role_key"
    ],
    errors="ignore"
)


# ------------------------------------------------------------
# 11. HANDLE MISSING RISK INFORMATION
# ------------------------------------------------------------

priority_candidate_df["joining_risk_score"] = (
    pd.to_numeric(
        priority_candidate_df[
            "joining_risk_score"
        ],
        errors="coerce"
    )
    .fillna(0)
)


priority_candidate_df["joining_risk_level"] = (
    priority_candidate_df[
        "joining_risk_level"
    ]
    .fillna("UNKNOWN")
    .astype(str)
    .str.upper()
    .str.strip()
)


priority_candidate_df["risk_reasons"] = (
    priority_candidate_df[
        "risk_reasons"
    ]
    .fillna(
        "Risk data unavailable for this candidate-role record"
    )
)


# ------------------------------------------------------------
# 12. COMBINED RECRUITER PRIORITY ACTION
# ------------------------------------------------------------
#
# Match intelligence alone answers:
# "Is this candidate suitable?"
#
# Joining-risk intelligence adds:
# "Does this candidate require urgent follow-up?"
#
# Together:
# "What should the recruiter do now?"
# ------------------------------------------------------------

def combined_action(row):

    match_level = (
        str(row["match_level"])
        .upper()
        .strip()
    )

    risk_level = (
        str(row["joining_risk_level"])
        .upper()
        .strip()
    )

    # Highest urgency:
    # Strong candidate + high joining risk
    if (
        match_level == "HIGH"
        and risk_level == "HIGH"
    ):
        return (
            "URGENT: High-fit candidate at joining risk "
            "— contact today"
        )

    # High-risk candidate requiring follow-up
    elif risk_level == "HIGH":
        return (
            "Existing candidate at joining risk "
            "— recruiter follow-up"
        )

    # Medium-risk high-match candidate
    elif (
        match_level == "HIGH"
        and risk_level == "MEDIUM"
    ):
        return (
            "High-fit candidate with moderate joining risk "
            "— follow up soon"
        )

    # Otherwise preserve original recommendation
    else:
        return row["recommended_action"]


priority_candidate_df[
    "combined_recruiter_action"
] = priority_candidate_df.apply(
    combined_action,
    axis=1
)


# ------------------------------------------------------------
# 13. SUMMARY
# ------------------------------------------------------------

print("\n========== FINAL RESULT ==========")

print(
    "Priority candidate-role rows:",
    len(priority_candidate_df)
)


print("\nJoining-risk coverage:")

print(
    priority_candidate_df[
        "joining_risk_level"
    ].value_counts()
)


# ------------------------------------------------------------
# 14. MOST URGENT CANDIDATES
# ------------------------------------------------------------

urgent_candidates = (
    priority_candidate_df[
        (
            priority_candidate_df[
                "match_level"
            ]
            .astype(str)
            .str.upper()
            .eq("HIGH")
        )
        &
        (
            priority_candidate_df[
                "joining_risk_level"
            ]
            .astype(str)
            .str.upper()
            .eq("HIGH")
        )
    ]
    [
        [
            "candidate_name",
            "target_role_display",
            "match_score",
            "joining_risk_score",
            "combined_recruiter_action"
        ]
    ]
    .sort_values(
        by=[
            "joining_risk_score",
            "match_score"
        ],
        ascending=False
    )
)


print(
    "\nHIGH-MATCH AND HIGH-RISK CANDIDATES "
    "(MOST URGENT):"
)


if len(urgent_candidates) > 0:

    display(
        urgent_candidates.head(15)
    )

else:

    print(
        "No HIGH-match + HIGH-risk candidates found."
    )


# ------------------------------------------------------------
# 15. FINAL CONFIDENTIALITY CHECK
# ------------------------------------------------------------

actual_client_names = [
    "Prowess",
    "Lodha Group",
    "Fischer Group",
    "L&T",
    "Leviat",
    "Hilti",
    "Alumil",
    "Lakshmi Hospital"
]


export_scan_text = (
    priority_candidate_df
    .astype(str)
    .agg(
        " ".join,
        axis=1
    )
    .str.cat(sep=" ")
)


detected_clients = [
    client
    for client in actual_client_names
    if client.lower()
    in export_scan_text.lower()
]


print(
    "\n========== CONFIDENTIALITY CHECK =========="
)


if detected_clients:

    print(
        "STATUS : EXPORT BLOCKED"
    )

    print(
        "Client names detected:",
        detected_clients
    )

    raise ValueError(
        "Confidential client information detected. "
        "Fix before exporting."
    )

else:

    print(
        "STATUS : CLEAN"
    )


# ------------------------------------------------------------
# 16. RE-EXPORT FINAL PRIORITY POOL
# ------------------------------------------------------------

export_path_csv = (
    "recruitment_priority_candidate_pool.csv"
)

export_path_xlsx = (
    "recruitment_priority_candidate_pool.xlsx"
)


priority_candidate_df.to_csv(
    export_path_csv,
    index=False
)


priority_candidate_df.to_excel(
    export_path_xlsx,
    index=False
)


print("\n========== EXPORT COMPLETE ==========")

print(
    f"{export_path_csv} "
    f"({len(priority_candidate_df):,} rows)"
)

print(
    f"{export_path_xlsx} "
    f"({len(priority_candidate_df):,} rows)"
)


print(
    "\nSTEP 25.3 COMPLETE — "
    "Joining-Risk Intelligence successfully integrated "
    "into the Recruiter Action Center."
)
```

```python
# ============================================================
# STEP 26 — RECRUITMENT INTELLIGENCE VISUALIZATION
# ============================================================
#
# Purpose:
# Management View of recruitment operations:
# - Recruitment volume
# - Role distribution
# - Match quality
# - Recruiter priority
# - High-priority opportunities
# - Joining-risk intelligence
# - Executive insights
#
# Display output is designed to use confidentiality-safe fields.
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re


print("=" * 70)
print("STEP 26 — RECRUITMENT INTELLIGENCE VISUALIZATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. INPUT VALIDATION
# ------------------------------------------------------------

required_dataframe_names = [
    "business_funnel_analysis_df",
    "client_performance_df",
    "role_effectiveness_df",
    "priority_candidate_df"
]

missing_dataframes = [
    name
    for name in required_dataframe_names
    if name not in globals()
]

if missing_dataframes:
    raise NameError(
        f"Missing required dataframe(s): {missing_dataframes}"
    )


print("\n========== INPUT VALIDATION ==========")

print(
    "Business funnel dataframe      :",
    business_funnel_analysis_df.shape
)

print(
    "Client performance dataframe   :",
    client_performance_df.shape
)

print(
    "Role effectiveness dataframe   :",
    role_effectiveness_df.shape
)

print(
    "Priority dataframe             :",
    priority_candidate_df.shape
)


# ------------------------------------------------------------
# 2. REQUIRED COLUMN VALIDATION
# ------------------------------------------------------------

required_priority_columns = [
    "candidate_name",
    "match_score",
    "match_level",
    "priority_category",
    "recommended_action",
    "joining_risk_score",
    "joining_risk_level",
    "combined_recruiter_action"
]

missing_priority_columns = [
    col
    for col in required_priority_columns
    if col not in priority_candidate_df.columns
]

if missing_priority_columns:
    raise ValueError(
        "priority_candidate_df is missing required columns: "
        f"{missing_priority_columns}"
    )


required_business_columns = [
    "candidate_name",
    "position_applied",
    "client_business"
]

missing_business_columns = [
    col
    for col in required_business_columns
    if col not in business_funnel_analysis_df.columns
]

if missing_business_columns:
    raise ValueError(
        "business_funnel_analysis_df is missing columns: "
        f"{missing_business_columns}"
    )


print("\n========== BASIC VALIDATION ==========")

print(
    "Recruitment population records :",
    len(business_funnel_analysis_df)
)

print(
    "Candidate-role match records   :",
    len(priority_candidate_df)
)

print(
    "Joining-risk integration       : AVAILABLE"
)


# ------------------------------------------------------------
# 3. CLIENT / BUSINESS GROUP DISTRIBUTION
# ------------------------------------------------------------

print(
    "\n========== BUSINESS GROUP DISTRIBUTION =========="
)


client_viz = (
    business_funnel_analysis_df
    .groupby("client_business")
    .size()
    .reset_index(name="candidate_count")
    .sort_values(
        "candidate_count",
        ascending=False
    )
)


print(
    client_viz.to_string(index=False)
)


plt.figure(figsize=(10, 6))

plt.barh(
    client_viz["client_business"],
    client_viz["candidate_count"]
)

plt.xlabel("Candidate Records")
plt.ylabel("Business Group")
plt.title("Recruitment Volume by Business Group")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 4. TOP RECRUITMENT ROLES
# ------------------------------------------------------------

print(
    "\n========== TOP RECRUITMENT ROLES =========="
)


required_role_columns = [
    "position_applied",
    "total_candidates"
]

missing_role_columns = [
    col
    for col in required_role_columns
    if col not in role_effectiveness_df.columns
]

if missing_role_columns:
    raise ValueError(
        "role_effectiveness_df is missing columns: "
        f"{missing_role_columns}"
    )


top_roles_viz = (
    role_effectiveness_df[
        [
            "position_applied",
            "total_candidates"
        ]
    ]
    .sort_values(
        "total_candidates",
        ascending=False
    )
    .head(10)
    .copy()
)


# ------------------------------------------------------------
# CONFIDENTIAL ROLE MASKING
# ------------------------------------------------------------

actual_client_names = [
    "Prowess",
    "Lodha Group",
    "Lodha",
    "Fischer Group",
    "Fischer",
    "L&T",
    "Leviat",
    "Hilti India",
    "Hilti",
    "Alumil",
    "Lakshmi Hospital"
]


def mask_client_names(text):

    if pd.isna(text):
        return text

    text = str(text)

    for client in actual_client_names:

        text = re.sub(
            rf"\b{re.escape(client)}\b",
            "",
            text,
            flags=re.IGNORECASE
        )

    text = re.sub(
        r"\s*[-–—]\s*",
        " ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


top_roles_viz[
    "confidential_role"
] = (
    top_roles_viz[
        "position_applied"
    ]
    .apply(mask_client_names)
)


print(
    top_roles_viz[
        [
            "confidential_role",
            "total_candidates"
        ]
    ]
    .to_string(index=False)
)


plt.figure(figsize=(10, 6))

plt.barh(
    top_roles_viz[
        "confidential_role"
    ],
    top_roles_viz[
        "total_candidates"
    ]
)

plt.xlabel("Candidate Records")
plt.ylabel("Role")
plt.title("Top Recruitment Roles by Candidate Volume")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 5. MATCH QUALITY DISTRIBUTION
# ------------------------------------------------------------

print(
    "\n========== MATCH QUALITY DISTRIBUTION =========="
)


match_quality_viz = (
    priority_candidate_df
    .groupby("match_level")
    .size()
    .reset_index(name="match_count")
    .sort_values(
        "match_count",
        ascending=False
    )
)


print(
    match_quality_viz.to_string(index=False)
)


plt.figure(figsize=(8, 5))

plt.bar(
    match_quality_viz["match_level"],
    match_quality_viz["match_count"]
)

plt.xlabel("Match Level")
plt.ylabel("Candidate-Role Matches")
plt.title("Candidate ↔ Role Match Quality")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. RECRUITER PRIORITY DISTRIBUTION
# ------------------------------------------------------------

print(
    "\n========== RECRUITER PRIORITY DISTRIBUTION =========="
)


priority_viz = (
    priority_candidate_df
    .groupby("priority_category")
    .size()
    .reset_index(name="match_count")
    .sort_values(
        "match_count",
        ascending=False
    )
)


print(
    priority_viz.to_string(index=False)
)


plt.figure(figsize=(10, 6))

plt.bar(
    priority_viz["priority_category"],
    priority_viz["match_count"]
)

plt.xlabel("Priority Category")
plt.ylabel("Candidate-Role Matches")
plt.title("Recruiter Priority Distribution")

plt.xticks(
    rotation=20,
    ha="right"
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7. JOINING-RISK DISTRIBUTION
# ------------------------------------------------------------

print(
    "\n========== JOINING-RISK DISTRIBUTION =========="
)


joining_risk_viz = (
    priority_candidate_df
    .groupby("joining_risk_level")
    .size()
    .reset_index(
        name="candidate_role_count"
    )
)


risk_order = {
    "HIGH": 1,
    "MEDIUM": 2,
    "LOW": 3,
    "UNKNOWN": 4
}


joining_risk_viz["risk_order"] = (
    joining_risk_viz[
        "joining_risk_level"
    ]
    .astype(str)
    .str.upper()
    .map(risk_order)
    .fillna(99)
)


joining_risk_viz = (
    joining_risk_viz
    .sort_values("risk_order")
    .drop(
        columns="risk_order"
    )
)


print(
    joining_risk_viz.to_string(index=False)
)


plt.figure(figsize=(8, 5))

plt.bar(
    joining_risk_viz[
        "joining_risk_level"
    ],
    joining_risk_viz[
        "candidate_role_count"
    ]
)

plt.xlabel("Joining-Risk Level")
plt.ylabel("Candidate-Role Matches")
plt.title("Joining-Risk Intelligence Distribution")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 8. HIGH-PRIORITY MATCHES BY BUSINESS GROUP
# ------------------------------------------------------------

print(
    "\n========== HIGH-PRIORITY MATCHES "
    "BY BUSINESS GROUP =========="
)


high_priority_categories = [
    "P1 - IMMEDIATE REVIEW",
    "P2 - HIGH PRIORITY"
]


high_priority_df = (
    priority_candidate_df[
        priority_candidate_df[
            "priority_category"
        ]
        .isin(high_priority_categories)
    ]
    .copy()
)


business_column = "confidential_client_business"


if business_column not in high_priority_df.columns:

    print(
        f"WARNING: {business_column} not available."
    )

    high_priority_business = pd.DataFrame(
        columns=[
            business_column,
            "high_priority_matches"
        ]
    )

else:

    high_priority_business = (
        high_priority_df
        .groupby(business_column)
        .size()
        .reset_index(
            name="high_priority_matches"
        )
        .sort_values(
            "high_priority_matches",
            ascending=False
        )
    )


if len(high_priority_business) > 0:

    print(
        high_priority_business.to_string(
            index=False
        )
    )

    plt.figure(figsize=(10, 6))

    plt.barh(
        high_priority_business[
            business_column
        ],
        high_priority_business[
            "high_priority_matches"
        ]
    )

    plt.xlabel("High-Priority Matches")
    plt.ylabel("Business Group")

    plt.title(
        "High-Priority Candidate Matches "
        "by Business Group"
    )

    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()

else:

    print(
        "No business-group data available "
        "for high-priority visualization."
    )


# ------------------------------------------------------------
# 9. REUSABLE TALENT PRIORITY
# ------------------------------------------------------------

print(
    "\n========== REUSABLE TALENT PRIORITY =========="
)


if (
    "in_reusable_pool"
    not in priority_candidate_df.columns
):

    print(
        "Reusable talent column not available."
    )

    reusable_priority_df = pd.DataFrame()

else:

    reusable_priority_df = (
        priority_candidate_df[
            (
                priority_candidate_df[
                    "in_reusable_pool"
                ]
                == True
            )
            &
            (
                priority_candidate_df[
                    "priority_category"
                ]
                .isin(
                    high_priority_categories
                )
            )
        ]
        .copy()
    )

    print(
        "Reusable high-priority matches :",
        len(reusable_priority_df)
    )


# ------------------------------------------------------------
# 10. TOP RECRUITER RECOMMENDATIONS
# ------------------------------------------------------------

print(
    "\n========== TOP 20 RECRUITER "
    "RECOMMENDATIONS =========="
)


recommendation_columns = [
    "candidate_name",
    "candidate_previous_role_display",
    "target_role_display",
    "match_score",
    "match_level",
    "priority_category",
    "joining_risk_level",
    "joining_risk_score",
    "combined_recruiter_action"
]


available_columns = [
    col
    for col in recommendation_columns
    if col in priority_candidate_df.columns
]


top_recommendations_df = (
    priority_candidate_df[
        priority_candidate_df[
            "priority_category"
        ]
        .isin(high_priority_categories)
    ]
    .sort_values(
        by=[
            "joining_risk_score",
            "match_score"
        ],
        ascending=[
            False,
            False
        ]
    )
    [available_columns]
    .head(20)
    .copy()
)


print(
    top_recommendations_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 11. HIGH-MATCH + HIGH-RISK URGENT CANDIDATES
# ------------------------------------------------------------

print(
    "\n========== URGENT FOLLOW-UP "
    "CANDIDATES =========="
)


urgent_candidates_viz = (
    priority_candidate_df[
        (
            priority_candidate_df[
                "match_level"
            ]
            .astype(str)
            .str.upper()
            .eq("HIGH")
        )
        &
        (
            priority_candidate_df[
                "joining_risk_level"
            ]
            .astype(str)
            .str.upper()
            .eq("HIGH")
        )
    ]
    .sort_values(
        by=[
            "joining_risk_score",
            "match_score"
        ],
        ascending=[
            False,
            False
        ]
    )
    .copy()
)


print(
    "High-match + High-risk candidates:",
    len(urgent_candidates_viz)
)


# ------------------------------------------------------------
# 12. EXECUTIVE INSIGHTS
# ------------------------------------------------------------

print(
    "\n========== EXECUTIVE INSIGHTS =========="
)


total_records = len(
    business_funnel_analysis_df
)


unique_candidates = (
    business_funnel_analysis_df[
        "candidate_name"
    ]
    .astype(str)
    .str.strip()
    .nunique()
)


unique_roles = (
    business_funnel_analysis_df[
        "position_applied"
    ]
    .nunique()
)


total_matches = len(
    priority_candidate_df
)


high_priority_count = len(
    high_priority_df
)


if total_matches > 0:

    high_priority_pct = (
        high_priority_count
        / total_matches
        * 100
    )

else:

    high_priority_pct = 0


high_risk_count = len(
    priority_candidate_df[
        priority_candidate_df[
            "joining_risk_level"
        ]
        .astype(str)
        .str.upper()
        .eq("HIGH")
    ]
)


if total_matches > 0:

    high_risk_pct = (
        high_risk_count
        / total_matches
        * 100
    )

else:

    high_risk_pct = 0


print(
    f"Total recruitment records       : "
    f"{total_records:,}"
)

print(
    f"Unique candidates               : "
    f"{unique_candidates:,}"
)

print(
    f"Unique recruitment roles        : "
    f"{unique_roles:,}"
)

print(
    f"Candidate-role matches          : "
    f"{total_matches:,}"
)

print(
    f"High-priority matches           : "
    f"{high_priority_count:,}"
)

print(
    f"High-priority match percentage  : "
    f"{high_priority_pct:.2f}%"
)

print(
    f"High joining-risk matches       : "
    f"{high_risk_count:,}"
)

print(
    f"High joining-risk percentage    : "
    f"{high_risk_pct:.2f}%"
)


if len(client_viz) > 0:

    print(
        "Highest recruitment volume      :",
        client_viz.iloc[0]["client_business"],
        "(",
        client_viz.iloc[0][
            "candidate_count"
        ],
        "records)"
    )


if len(top_roles_viz) > 0:

    print(
        "Highest-volume role             :",
        top_roles_viz.iloc[0][
            "confidential_role"
        ],
        "(",
        top_roles_viz.iloc[0][
            "total_candidates"
        ],
        "records)"
    )


if "date" in business_funnel_analysis_df.columns:

    print(
        "Active recruitment dates        :",
        business_funnel_analysis_df[
            "date"
        ].nunique()
    )

    print(
        "Recruitment data period         :",
        business_funnel_analysis_df[
            "date"
        ].min(),
        "→",
        business_funnel_analysis_df[
            "date"
        ].max()
    )


# ------------------------------------------------------------
# 13. FINAL CONFIDENTIALITY CHECK
# ------------------------------------------------------------

print(
    "\n========== CONFIDENTIALITY CHECK =========="
)


display_columns_to_scan = [
    "candidate_previous_role_display",
    "target_role_display",
    "candidate_status"
]


detected_clients = []


for col in display_columns_to_scan:

    if col not in priority_candidate_df.columns:
        continue

    text = (
        priority_candidate_df[
            col
        ]
        .astype(str)
        .str.cat(sep=" ")
    )

    for client in actual_client_names:

        if client.lower() in text.lower():

            detected_clients.append(
                client
            )


detected_clients = sorted(
    set(detected_clients)
)


print(
    "Actual client names detected :",
    detected_clients
)


if detected_clients:

    raise ValueError(
        "CONFIDENTIALITY CHECK FAILED. "
        f"Detected client names: "
        f"{detected_clients}"
    )


print(
    "STATUS : CLEAN"
)


# ------------------------------------------------------------
# 14. FINAL VALIDATION
# ------------------------------------------------------------

print(
    "\n========== STEP 26 VALIDATION =========="
)


print(
    "Recruitment population available :",
    len(business_funnel_analysis_df) > 0
)

print(
    "Priority records available       :",
    len(priority_candidate_df) > 0
)

print(
    "Business groups represented      :",
    business_funnel_analysis_df[
        "client_business"
    ].nunique()
)

print(
    "Top-role visualization created   :",
    len(top_roles_viz) > 0
)

print(
    "High-priority visualization created:",
    len(high_priority_business) > 0
)

print(
    "Joining-risk visualization created:",
    len(joining_risk_viz) > 0
)

print(
    "Executive insights generated     :",
    True
)

print(
    "Confidential display checked     :",
    len(detected_clients) == 0
)


print("\n" + "=" * 70)
print("STEP 26 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 30 — RECRUITMENT INTELLIGENCE DASHBOARD
# DYNAMIC CSV/XLSX UPLOAD VERSION — WITH JOINING-RISK + TOKEN-SAFE DETECTION
# ================================================================

print("=" * 70)
print("RECRUITMENT INTELLIGENCE DASHBOARD")
print("=" * 70)

import pandas as pd
import numpy as np
import gradio as gr
import matplotlib.pyplot as plt
import os
import re


# ============================================================
# 1. CONFIDENTIALITY / CLIENT MASKING
# ============================================================

# FIXED: expanded from 7 to 11 keys, matching the fuller list used
# elsewhere in the notebook (Step 27's actual_client_names). The
# original map here was missing "Alumil" entirely and bare "Lodha"
# / "Fischer" — since this dashboard is meant to accept the
# project's own raw/pre-standardized data as a valid upload, an
# "Alumil" record would have displayed unmasked in the live chart.
CLIENT_MAP = {
    "Prowess": "Technology & Business Consulting",
    "Lodha Group": "Real Estate & Infrastructure",
    "Lodha": "Real Estate & Infrastructure",
    "Fischer Group": "Industrial & Engineering Solutions",
    "Fischer": "Industrial & Engineering Solutions",
    "L&T": "Engineering & Infrastructure",
    "Leviat": "Construction Products & Solutions",
    "Alumil": "Construction Products & Solutions",
    "Hilti India": "Building & Industrial Solutions",
    "Hilti": "Building & Industrial Solutions",
    "Lakshmi Hospital": "Healthcare & Hospital Services"
}


def mask_client(value):
    if pd.isna(value):
        return value
    text = str(value)
    for actual, masked in CLIENT_MAP.items():
        if actual.lower() in text.lower():
            return masked
    return text


def _empty_chart(message):
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.axis("off")
    ax.text(
        0.5, 0.5, message,
        ha="center", va="center",
        fontsize=12, color="#6b7280", wrap=True
    )
    return fig


# ============================================================
# 2. COLUMN DETECTION — TOKEN-BOUNDARY MATCHING
#    (fixes false positives like "name" matching inside
#    "educational_institution_name", or "role" matching inside
#    "role_positions" — same bug found and fixed in STEP 31)
# ============================================================

def _tokens(s):
    return set(str(s).strip().lower().replace(" ", "_").split("_"))


def detect_column(df, candidates, claimed=None):
    if df is None:
        return None
    if claimed is None:
        claimed = set()

    lookup = {str(c).strip().lower(): c for c in df.columns}

    # Exact match first
    for candidate in candidates:
        key = candidate.strip().lower()
        if key in lookup and lookup[key] not in claimed:
            return lookup[key]

    # Whole-word/phrase match only — an alias must be a complete token
    # inside the column name, not just a substring.
    for column in df.columns:
        if column in claimed:
            continue
        col_tokens = _tokens(column)
        for candidate in candidates:
            cand_tokens = _tokens(candidate)
            if cand_tokens.issubset(col_tokens):
                return column

    return None


CANDIDATE_ALIASES = [
    "candidate_name", "candidate name", "full_name", "full name",
    "applicant_name", "applicant name"
]  # dropped bare "candidate", "name", "applicant"

ROLE_ALIASES = [
    "target_role", "target role", "candidate_previous_role", "position_applied",
    "position applied", "applied for", "applied_for", "job role", "job title",
    "job_title", "designation", "recruitment_role", "positions"
]  # dropped bare "role", "position", "title"

BUSINESS_ALIASES = [
    "confidential_client_business", "client_business", "business_group",
    "business group", "client company", "client_company", "company name",
    "company_name", "client name", "client_name"
]  # dropped bare "client", "company"

MATCH_LEVEL_ALIASES = ["match_level", "match level"]
MATCH_SCORE_ALIASES = ["match_score", "match score"]
PRIORITY_ALIASES = ["priority_category", "priority category", "priority"]
RISK_LEVEL_ALIASES = ["joining_risk_level", "joining risk level", "risk_level"]
RISK_SCORE_ALIASES = ["joining_risk_score", "joining risk score", "risk_score"]
ACTION_ALIASES = ["combined_recruiter_action", "recommended_action", "recommended action"]


# ============================================================
# 3. LOAD FILE
# ============================================================

def load_uploaded_file(file):
    if file is None:
        return None, "Please upload a CSV or XLSX file."
    try:
        file_path = file.name
        extension = os.path.splitext(file_path)[1].lower()

        if extension == ".csv":
            df = pd.read_csv(file_path)
        elif extension in [".xlsx", ".xls"]:
            df = pd.read_excel(file_path)
        else:
            return None, "Unsupported file type. Please upload CSV or XLSX."

        df = df.dropna(how="all").copy()
        df.columns = [str(c).strip() for c in df.columns]

        if df.empty:
            return None, "The uploaded file is empty."

        return df, (
            f"File loaded successfully: "
            f"{len(df):,} records x {len(df.columns)} columns"
        )

    except Exception as e:
        return None, f"Unable to read file: {str(e)}"


# ============================================================
# 4. RECRUITMENT ANALYSIS
# ============================================================

def analyze_recruitment(file):

    df, message = load_uploaded_file(file)

    if df is None:
        # IMPORTANT: this tuple must have exactly the same length
        # and order as the successful-analysis return below (18
        # values) -- it used to have 19, one more than the 18
        # Gradio output components, which crashes the UI the
        # moment someone uploads a file that can't be read (wrong
        # format, corrupted, truly empty). That is a real
        # "unknown data" case, not an edge case to ignore.
        blank_chart = _empty_chart(message)
        return (
            message,
            0, 0, 0, 0,
            0, 0, 0, 0,
            pd.DataFrame(),
            blank_chart,
            pd.DataFrame(),
            pd.DataFrame(),
            blank_chart,
            pd.DataFrame(),
            blank_chart,
            pd.DataFrame(),
            pd.DataFrame()
        )

    # --------------------------------------------------------
    # Detect important columns (candidate/role/business share a
    # "claimed" set so one column can't be double-assigned)
    # --------------------------------------------------------

    claimed_columns = set()

    candidate_col = detect_column(df, CANDIDATE_ALIASES, claimed_columns)
    if candidate_col:
        claimed_columns.add(candidate_col)

    role_col = detect_column(df, ROLE_ALIASES, claimed_columns)
    if role_col:
        claimed_columns.add(role_col)

    business_col = detect_column(df, BUSINESS_ALIASES, claimed_columns)
    if business_col:
        claimed_columns.add(business_col)

    match_level_col = detect_column(df, MATCH_LEVEL_ALIASES)
    priority_col = detect_column(df, PRIORITY_ALIASES)
    risk_level_col = detect_column(df, RISK_LEVEL_ALIASES)
    risk_score_col = detect_column(df, RISK_SCORE_ALIASES)
    action_col = detect_column(df, ACTION_ALIASES)

    # ========================================================
    # 5. BASIC KPIs
    # ========================================================

    total_records = len(df)

    unique_candidates = (
        df[candidate_col].dropna().astype(str).str.strip().nunique()
        if candidate_col else 0
    )

    unique_roles = (
        df[role_col].dropna().astype(str).str.strip().nunique()
        if role_col else 0
    )

    # NOTE: this is a simplified live heuristic — any candidate who
    # appears against more than one role in THIS uploaded file — not
    # the same definition as the offline reusable_talent_pool_df
    # (267 candidates) built from Step 22's selection-signal logic.
    # They will not match if you upload the project's own data here.
    reusable_candidates = 0
    if candidate_col and role_col:
        temp = df[[candidate_col, role_col]].dropna()
        reusable_counts = temp.groupby(candidate_col)[role_col].nunique()
        reusable_candidates = (reusable_counts > 1).sum()

    # ========================================================
    # 6. MATCH ANALYSIS
    # ========================================================

    match_count = 0
    high_matches = 0
    high_priority_matches = 0
    high_risk_matches = 0

    match_summary = pd.DataFrame(columns=["Match Level", "Matches"])
    priority_summary = pd.DataFrame(columns=["Priority Category", "Matches"])
    risk_summary = pd.DataFrame(columns=["Joining Risk Level", "Candidate-Role Rows"])

    if match_level_col:
        levels = df[match_level_col].fillna("NO MATCH").astype(str).str.upper().str.strip()
        match_count = len(df)
        match_summary = levels.value_counts().rename_axis("Match Level").reset_index(name="Matches")
        high_matches = (levels == "HIGH").sum()

    # ========================================================
    # 7. PRIORITY ANALYSIS
    # ========================================================

    if priority_col:
        priorities = df[priority_col].fillna("P4 - LOW / NO PRIORITY").astype(str).str.upper().str.strip()
        priority_summary = priorities.value_counts().rename_axis("Priority Category").reset_index(name="Matches")
        high_priority_matches = priorities.str.contains("P1|P2|HIGH", case=False, regex=True).sum()

    # ========================================================
    # 8. JOINING-RISK ANALYSIS
    # ========================================================

    if risk_level_col:
        risk_levels = df[risk_level_col].fillna("UNKNOWN").astype(str).str.upper().str.strip()
        risk_summary = risk_levels.value_counts().rename_axis("Joining Risk Level").reset_index(name="Candidate-Role Rows")
        high_risk_matches = (risk_levels == "HIGH").sum()

    # ========================================================
    # 9. BUSINESS GROUP ANALYSIS
    # ========================================================

    business_summary = pd.DataFrame(columns=["Business Group", "Candidate Records"])
    business_plot = None

    if business_col:
        business_data = df.copy()
        business_data["Business Group"] = business_data[business_col].apply(mask_client)
        business_summary = business_data["Business Group"].fillna("Not Available").value_counts().reset_index()
        business_summary.columns = ["Business Group", "Candidate Records"]

        chart_data = business_summary.head(10)
        fig, ax = plt.subplots(figsize=(9, 5))
        ax.barh(chart_data["Business Group"][::-1], chart_data["Candidate Records"][::-1])
        ax.set_title("Recruitment by Business Group")
        ax.set_xlabel("Candidate Records")
        plt.tight_layout()
        business_plot = fig
    else:
        business_plot = _empty_chart("No client/business column found in this file.")

    # ========================================================
    # 10. TOP ROLES
    # ========================================================

    role_summary = pd.DataFrame(columns=["Role", "Candidate Records"])
    if role_col:
        role_summary = df[role_col].fillna("Unknown").astype(str).str.strip().value_counts().head(10).reset_index()
        role_summary.columns = ["Role", "Candidate Records"]

    # ========================================================
    # 11. MATCH QUALITY CHART
    # ========================================================

    match_plot = None
    if not match_summary.empty:
        fig, ax = plt.subplots(figsize=(8, 5))
        ax.bar(match_summary["Match Level"], match_summary["Matches"])
        ax.set_title("Candidate-Role Match Quality")
        ax.set_ylabel("Matches")
        plt.xticks(rotation=20)
        plt.tight_layout()
        match_plot = fig
    else:
        match_plot = _empty_chart("No match_level column found -- this file hasn't been through JD matching.")

    # ========================================================
    # 12. JOINING-RISK CHART
    # ========================================================

    risk_plot = None
    if not risk_summary.empty:
        order = ["HIGH", "MEDIUM", "LOW", "UNKNOWN"]
        risk_summary["Joining Risk Level"] = pd.Categorical(risk_summary["Joining Risk Level"], categories=order, ordered=True)
        risk_summary = risk_summary.sort_values("Joining Risk Level")

        fig, ax = plt.subplots(figsize=(8, 5))
        colors = {"HIGH": "#dc2626", "MEDIUM": "#f59e0b", "LOW": "#16a34a", "UNKNOWN": "#94a3b8"}
        bar_colors = [colors.get(str(lvl), "#94a3b8") for lvl in risk_summary["Joining Risk Level"]]
        ax.bar(risk_summary["Joining Risk Level"].astype(str), risk_summary["Candidate-Role Rows"], color=bar_colors)
        ax.set_title("Joining-Risk Distribution")
        ax.set_ylabel("Candidate-Role Rows")
        plt.tight_layout()
        risk_plot = fig
    else:
        risk_plot = _empty_chart("No joining_risk_level column found in this file.")

    # ========================================================
    # 13. CANDIDATE RECOMMENDATIONS
    #     Reuses the SAME columns already detected above (
    #     candidate_col, role_col, match_level_col, etc.) instead
    #     of re-detecting from a separate narrower alias list --
    #     so any column found for the KPIs/charts always shows up
    #     here too. (This fixes a real bug: "Applicant Name" was
    #     detected fine for the KPI cards but used to be dropped
    #     from this table because it wasn't re-detected here.)
    # ========================================================

    match_score_col = detect_column(df, MATCH_SCORE_ALIASES)

    rec_column_map = []  # (source_column_in_df, display_name)
    if candidate_col:
        rec_column_map.append((candidate_col, "Candidate"))
    if role_col:
        rec_column_map.append((role_col, "Target Role"))
    if match_score_col:
        rec_column_map.append((match_score_col, "Match Score"))
    if match_level_col:
        rec_column_map.append((match_level_col, "Match Level"))
    if risk_score_col:
        rec_column_map.append((risk_score_col, "Joining Risk Score"))
    if risk_level_col:
        rec_column_map.append((risk_level_col, "Joining Risk Level"))
    if priority_col:
        rec_column_map.append((priority_col, "Priority"))
    if action_col:
        rec_column_map.append((action_col, "Recommended Action"))

    # de-duplicate by source column, keep first occurrence
    seen_cols = set()
    rec_column_map = [
        pair for pair in rec_column_map
        if not (pair[0] in seen_cols or seen_cols.add(pair[0]))
    ]

    if rec_column_map:
        source_cols = [c for c, _ in rec_column_map]
        recommendations = df[source_cols].copy()
        recommendations.columns = [name for _, name in rec_column_map]

        sort_cols = [c for c in ["Match Score", "Joining Risk Score"] if c in recommendations.columns]
        if sort_cols:
            for c in sort_cols:
                recommendations[c] = pd.to_numeric(recommendations[c], errors="coerce")
            recommendations = recommendations.sort_values(sort_cols, ascending=False, na_position="last")

        recommendations = recommendations.head(50)
    else:
        recommendations = pd.DataFrame({
            "Message": ["Candidate recommendation fields were not found in the uploaded file."]
        })

    # ========================================================
    # 14. STATUS MESSAGE — reports what was detected AND what
    #     is missing, instead of silently showing zeros.
    # ========================================================

    detected = []
    missing_notes = []

    if candidate_col:
        detected.append("Candidate")
    else:
        missing_notes.append("No candidate-name column found -> candidate counts will show 0.")

    if role_col:
        detected.append("Role")
    else:
        missing_notes.append("No role/position column found -> Unique Roles, Top Roles and Reusable Candidates will show 0/empty.")

    if business_col:
        detected.append("Business")
    else:
        missing_notes.append("No client/business column found -> Business Group section will be empty.")

    if match_level_col:
        detected.append("Match")
    else:
        missing_notes.append("No match_level column found -> this file has not been through the JD-matching step, so Match Quality will be empty.")

    if priority_col:
        detected.append("Priority")
    else:
        missing_notes.append("No priority_category column found -> Recruiter Prioritization will be empty.")

    if risk_level_col:
        detected.append("Joining Risk")
    else:
        missing_notes.append("No joining_risk_level column found -> Joining-Risk Overview will be empty.")

    status_lines = [
        "Analysis completed successfully.",
        "",
        f"Records processed: {total_records:,}",
        f"Columns detected: {', '.join(detected) if detected else 'Basic dataset only'}",
    ]

    if missing_notes:
        status_lines.append("")
        status_lines.append("Notes on this file (not errors -- just what could not be found):")
        for note in missing_notes:
            status_lines.append(f"- {note}")

    status_message = "\n".join(status_lines)

    return (
        status_message,
        total_records, unique_candidates, unique_roles, reusable_candidates,
        match_count, high_matches, high_priority_matches, high_risk_matches,
        business_summary, business_plot,
        role_summary,
        match_summary, match_plot,
        risk_summary, risk_plot,
        priority_summary,
        recommendations
    )


# ============================================================
# 15. GRADIO UI
# ============================================================

with gr.Blocks(title="Recruitment Intelligence Dashboard") as demo:

    gr.Markdown("""
    # Recruitment Intelligence Dashboard

    **Analyze recruitment data, candidate-role matching, recruiter priorities
    and joining-risk from an uploaded CSV or XLSX file.**
    """)

    with gr.Row():
        file_input = gr.File(label="Upload Recruitment Data", file_types=[".csv", ".xlsx", ".xls"], type="filepath")
        analyze_button = gr.Button("Analyze Recruitment Data", variant="primary")

    status_box = gr.Markdown("Upload a recruitment CSV/XLSX file and click **Analyze Recruitment Data**.")

    gr.Markdown("## Recruitment Overview")
    with gr.Row():
        kpi1 = gr.Number(label="Recruitment Records", value=0, interactive=False)
        kpi2 = gr.Number(label="Unique Candidates", value=0, interactive=False)
        kpi3 = gr.Number(label="Unique Roles", value=0, interactive=False)
        kpi4 = gr.Number(label="Reusable Candidates", value=0, interactive=False)

    gr.Markdown("## Candidate-Role Matching")
    with gr.Row():
        kpi5 = gr.Number(label="Candidate-Role Matches", value=0, interactive=False)
        kpi6 = gr.Number(label="HIGH Matches", value=0, interactive=False)
        kpi7 = gr.Number(label="High-Priority Matches", value=0, interactive=False)
        kpi8 = gr.Number(label="High Joining-Risk Rows", value=0, interactive=False)

    gr.Markdown("## Recruitment by Business Group")
    business_table = gr.Dataframe(headers=["Business Group", "Candidate Records"], interactive=False)
    business_plot_component = gr.Plot()

    gr.Markdown("## Top Recruitment Roles")
    role_table = gr.Dataframe(headers=["Role", "Candidate Records"], interactive=False)

    gr.Markdown("## Match Quality")
    match_table = gr.Dataframe(headers=["Match Level", "Matches"], interactive=False)
    match_plot_component = gr.Plot()

    gr.Markdown("## Joining-Risk Overview")
    risk_table = gr.Dataframe(headers=["Joining Risk Level", "Candidate-Role Rows"], interactive=False)
    risk_plot_component = gr.Plot()

    gr.Markdown("## Recruiter Prioritization")
    priority_table = gr.Dataframe(headers=["Priority Category", "Matches"], interactive=False)

    gr.Markdown("## Top Candidate Recommendations")
    recommendation_table = gr.Dataframe(interactive=False)

    analyze_button.click(
        fn=analyze_recruitment,
        inputs=file_input,
        outputs=[
            status_box,
            kpi1, kpi2, kpi3, kpi4,
            kpi5, kpi6, kpi7, kpi8,
            business_table, business_plot_component,
            role_table,
            match_table, match_plot_component,
            risk_table, risk_plot_component,
            priority_table,
            recommendation_table
        ]
    )

print("\nLaunching Recruitment Intelligence Dashboard...")
demo.launch(share=False, debug=True, inline=True, show_error=True)
```

```python
# ============================================================
# STEP 30 (FALLBACK) — STATIC IN-NOTEBOOK DASHBOARD
# No Gradio server, no tunnel, no browser dependency.
# Runs analyze_recruitment() directly and renders results
# as notebook output: tables + charts, same logic as the
# Gradio version above.
# ============================================================

from IPython.display import display, Markdown

def run_dashboard(file_path):
    class FakeUpload:
        def __init__(self, path):
            self.name = path

    result = analyze_recruitment(FakeUpload(file_path))

    (
        status_message,
        total_records, unique_candidates, unique_roles, reusable_candidates,
        match_count, high_matches, high_priority_matches, high_risk_matches,
        business_summary, business_plot,
        role_summary,
        match_summary, match_plot,
        risk_summary, risk_plot,
        priority_summary,
        recommendations
    ) = result

    display(Markdown("# Recruitment Intelligence Dashboard"))
    display(Markdown(f"**{status_message}**".replace("\n", "  \n")))

    display(Markdown("## Recruitment Overview"))
    print(f"Recruitment Records   : {total_records:,}")
    print(f"Unique Candidates     : {unique_candidates:,}")
    print(f"Unique Roles          : {unique_roles:,}")
    print(f"Reusable Candidates   : {reusable_candidates:,}")

    display(Markdown("## Candidate-Role Matching"))
    print(f"Candidate-Role Matches : {match_count:,}")
    print(f"HIGH Matches           : {high_matches:,}")
    print(f"High-Priority Matches  : {high_priority_matches:,}")
    print(f"High Joining-Risk Rows : {high_risk_matches:,}")

    display(Markdown("## Recruitment by Business Group"))
    display(business_summary)
    if business_plot:
        business_plot.show()

    display(Markdown("## Top Recruitment Roles"))
    display(role_summary)

    display(Markdown("## Match Quality"))
    display(match_summary)
    if match_plot:
        match_plot.show()

    display(Markdown("## Joining-Risk Overview"))
    display(risk_summary)
    if risk_plot:
        risk_plot.show()

    display(Markdown("## Recruiter Prioritization"))
    display(priority_summary)

    display(Markdown("## Top Candidate Recommendations"))
    display(recommendations.head(20))

    return result


# ============================================================
# RUN ON YOUR OWN 800-RECORD DATASET FOR THE PRESENTATION
# ============================================================

result = run_dashboard("/content/recruitment_priority_candidate_pool.csv")  # <-- set path
```
